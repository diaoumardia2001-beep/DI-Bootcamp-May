```python
import platform
import tensorflow as tf
import tensorflow_datasets as tfds
from transformers import BertTokenizer, TFBertForSequenceClassification

print("Python version      :", platform.python_version())
print("TensorFlow version  :", tf.__version__)
print("GPU devices detected:", tf.config.list_physical_devices('GPU'))
```

```python
(ds_train, ds_test), ds_info = tfds.load(
    "imdb_reviews",
    split=(tfds.Split.TRAIN, tfds.Split.TEST),
    as_supervised=True,
    with_info=True
)
print(ds_info)
```

```python
for text, label in ds_train.take(2):
    print("Label:", "Positive" if label.numpy() else "Negative")
    print(text.numpy().decode()[:250], "...\n")
```

```python
MAX_LENGTH = 256
BATCH_SIZE = 16

tokenizer = BertTokenizer.from_pretrained("bert-base-uncased", do_lower_case=True)
print("Tokenizer loaded:", tokenizer.name_or_path)
```

```python
def encode_review(review_input):
    if isinstance(review_input, bytes):
        review_text = review_input.decode("utf-8")
    elif hasattr(review_input, "numpy"):
        review_text = review_input.numpy().decode("utf-8")
    else:
        review_text = str(review_input)

    return tokenizer.encode_plus(
        review_text,
        add_special_tokens=True,
        max_length=MAX_LENGTH,
        padding="max_length",
        truncation=True,
        return_attention_mask=True,
        return_token_type_ids=True,
    )

def tf_encode(text, label):
    encoded = tf.py_function(
        func=lambda t: list(encode_review(t).values()),
        inp=[text],
        Tout=[tf.int32, tf.int32, tf.int32]
    )
    return {
        "input_ids": encoded[0],
        "attention_mask": encoded[1],
        "token_type_ids": encoded[2]
    }, label

def prepare_dataset(dataset):
    return (
        dataset
        .map(tf_encode, num_parallel_calls=tf.data.AUTOTUNE)
        .shuffle(2000)
        .batch(BATCH_SIZE)
        .prefetch(tf.data.AUTOTUNE)
    )

train_ds = prepare_dataset(ds_train)
test_ds  = prepare_dataset(ds_test)
```

```python
model = TFBertForSequenceClassification.from_pretrained(
    "bert-base-uncased",
    num_labels=2,
    use_safetensors=False
)

optimizer = tf.keras.optimizers.Adam(learning_rate=2e-5, epsilon=1e-8)
loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
metrics = [tf.keras.metrics.SparseCategoricalAccuracy(name="accuracy")]

model.compile(optimizer=optimizer, loss=loss_fn, metrics=metrics)
model.summary()
```

```python
EPOCHS = 2

history = model.fit(
    train_ds,
    validation_data=test_ds,
    epochs=EPOCHS
)
```

```python
eval_metrics = model.evaluate(test_ds, return_dict=True)
print(eval_metrics)
```

```python
def predict_sentiment(text: str):
    encoded = tokenizer.encode_plus(
        text,
        add_special_tokens=True,
        max_length=MAX_LENGTH,
        padding="max_length",
        truncation=True,
        return_attention_mask=True,
        return_token_type_ids=True,
        return_tensors="tf"
    )

    inputs = {
        "input_ids": encoded["input_ids"],
        "attention_mask": encoded["attention_mask"],
        "token_type_ids": encoded["token_type_ids"]
    }

    logits = model(inputs).logits
    probs = tf.nn.softmax(logits, axis=-1).numpy()[0]
    predicted_class = int(probs.argmax())
    label = "Positive" if predicted_class == 1 else "Negative"

    return label, float(probs.max())

custom_sentence = "The onboarding emails were confusing, but the agent fixed everything politely."
label, confidence = predict_sentiment(custom_sentence)
print(f"Prediction: {label} (confidence={confidence:.3f})")
```

1. Which lever most improved results?
Sequence length (MAX_LENGTH) and learning rate tend to matter most in a setup like this. Increasing MAX_LENGTH beyond 256 tokens usually helps most with IMDB reviews specifically, since many reviews are long and get truncated at 256 — you lose information from the end of the review, which sometimes contains the strongest sentiment signal ("but overall, I loved it"). A learning rate in the 2e-5–3e-5 range is close to optimal for fine-tuning BERT; going higher tends to destabilize training (loss spikes), while going lower slows convergence without much accuracy gain. Increasing epochs from 2 to 3 usually gives a smaller improvement and risks overfitting, since BERT converges fast on a well-balanced dataset like IMDB.

2. Where would you add guardrails before deploying this live?
A few guardrails are essential before production use:

Confidence thresholding: route predictions below a certain confidence (e.g. <0.75) to a human reviewer instead of auto-responding, since low-confidence predictions are more error-prone.
Domain mismatch monitoring: this model was fine-tuned on movie reviews, not support tickets. Language, tone, and vocabulary differ (technical jargon, product names, sarcasm), so accuracy in production will likely be lower than the ~90% benchmark seen on IMDB. A guardrail here is running a small labeled sample of real support tickets through the model before full deployment to measure the actual gap.
Drift detection: track the distribution of predicted labels and confidence scores over time; a sudden shift can signal that customer language has changed (new product issue, new slang) and the model needs retraining.
Bias and edge-case testing: check performance on short messages, non-native English phrasing, and messages mixing positive and negative sentiment (mixed reviews), which are common failure modes for sentiment classifiers.
Human-in-the-loop escalation path: any negative sentiment above a severity threshold (repeated complaints, cancellation language) should always be flagged for human review regardless of model confidence, since the cost of missing a churn signal is high.
