```python
from datasets import load_dataset
import pandas as pd

# 1. Chargement sans erreur avec le bon namespace
dataset = load_dataset("cardiffnlp/tweet_eval", "sentiment")
print("Structure du dataset :", dataset)

# 2. Analyse de la répartition (0: Négatif, 1: Neutre, 2: Positif)
df_train = pd.DataFrame(dataset["train"])
print("\nRépartition des classes (Train) :")
print(df_train["label"].value_counts(normalize=True))

# 3. Sauvegarde de deux exemples par classe pour l'Étape 6
examples = {}
for label_id, label_name in enumerate(["Negative", "Neutral", "Positive"]):
    subset = df_train[df_train["label"] == label_id].head(2)
    examples[label_name] = subset["text"].tolist()

print("\nExemples mis de côté :", examples)
```

```python
from transformers import AutoTokenizer

MODEL_NAME = "distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def preprocess_function(examples):
    # Les tweets sont courts : 128 tokens suffisent largement
    return tokenizer(
        examples["text"],
        truncation=True,
        padding="max_length",
        max_length=128
    )

# Application industrielle en mode "batched"
tokenized_datasets = dataset.map(preprocess_function, batched=True)
tokenized_datasets = tokenized_datasets.remove_columns(["text"])
tokenized_datasets.set_format("torch")
```

```python
import numpy as np
import evaluate
from transformers import AutoModelForSequenceClassification, TrainingArguments, Trainer

# 1. Métriques d'évaluation
accuracy_metric = evaluate.load("accuracy")
f1_metric = evaluate.load("f1")

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    acc = accuracy_metric.compute(predictions=predictions, references=labels)["accuracy"]
    f1 = f1_metric.compute(predictions=predictions, references=labels, average="macro")["f1"]
    return {"accuracy": acc, "macro_f1": f1}

# 2. Initialisation du modèle (3 classes)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=3)

# 3. Arguments d'entraînement
training_args = TrainingArguments(
    output_dir="./results_distilbert_sentiment",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=5e-5,
    per_device_train_batch_size=32,
    per_device_eval_batch_size=32,
    num_train_epochs=3,
    weight_decay=0.01,
    load_best_model_at_end=True,
    metric_for_best_model="macro_f1",
    logging_steps=100,
    report_to="none" # Évite les connexions tierces (WandB/TensorBoard) pour ce run local
)

# 4. Instanciation du Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["validation"],
    compute_metrics=compute_metrics,
)

# Lancement de l'entraînement
trainer.train()
trainer.save_model("./best_sentiment_model")
tokenizer.save_pretrained("./best_sentiment_model")
```

```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt

# 1. Évaluation standard sur le Validation Split
val_results = trainer.evaluate(tokenized_datasets["validation"])
print("\nRésultats Validation :", val_results)

# 2. Récupération des scores de confiance sur le Test Split
predictions_output = trainer.predict(tokenized_datasets["test"])
logits = torch.tensor(predictions_output.predictions)
probabilities = F.softmax(logits, dim=-1).numpy()

# Confiance associée à la classe prédite
confidence_scores = np.max(probabilities, axis=-1)

# 3. Histogramme de calibration
plt.figure(figsize=(8, 5))
plt.hist(confidence_scores, bins=np.arange(0.3, 1.1, 0.1), edgecolor='black', color='#4F46E5', rwidth=0.8)
plt.title("Distribution des Scores de Confiance (Calibration)")
plt.xlabel("Score de Confiance (Max Softmax)")
plt.ylabel("Nombre de Tweets")
plt.grid(axis='y', alpha=0.5)
plt.show()
```

```python
import torch
import seaborn as sns

# Charger le modèle pour l'extraction d'attention
model_explain = AutoModelForSequenceClassification.from_pretrained(
    "./best_sentiment_model",
    output_attentions=True
)
model_explain.eval()

def analyze_and_visualize_attention(text_sample):
    inputs = tokenizer(text_sample, return_tensors="pt", truncation=True, max_length=128)
    tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

    with torch.no_grad():
        outputs = model_explain(**inputs)

    # logits et probabilités
    probs = torch.softmax(outputs.logits, dim=-1).numpy()[0]
    pred_label = np.argmax(probs)
    labels_mapping = {0: "Negative", 1: "Neutral", 2: "Positive"}

    # Récupération des attentions : tuple de taille (num_layers)
    # Chaque élément a la forme : (batch, num_heads, sequence_length, sequence_length)
    attentions = outputs.attentions
    last_layer_attention = attentions[-1][0] # Dernière couche, premier item du batch

    # Moyenne sur l'ensemble des têtes d'attention (Attention Heads)
    avg_attention = torch.mean(last_layer_attention, dim=0)

    # Extraction de l'attention portée par le token spécial [CLS] (index 0) sur tous les autres tokens
    cls_attention = avg_attention[0].numpy()

    # Formatage pour l'affichage (Top 5 tokens clés hors tokens spéciaux)
    token_attention_pairs = list(zip(tokens, cls_attention))
    filtered_pairs = [p for p in token_attention_pairs if p[0] not in ['[CLS]', '[SEP]', '[PAD]']]
    top_tokens = sorted(filtered_pairs, key=lambda x: x[1], reverse=True)[:5]

    # Visualisation sous forme de Bar Chart
    plt.figure(figsize=(10, 3))
    sns.barplot(x=[t[0] for t in filtered_pairs], y=[t[1] for t in filtered_pairs], color="#0EA5E9")
    plt.xticks(rotation=45)
    plt.title(f"Tokens ciblés par le jeton [CLS] | Prédiction : {labels_mapping[pred_label]} ({probs[pred_label]:.2f})")
    plt.ylabel("Poids d'attention moyen")
    plt.show()

    return {
        "label": labels_mapping[pred_label],
        "confidence": float(probs[pred_label]),
        "highlighted_tokens": top_tokens
    }

# Test sur l'un de vos exemples négatifs sauvegardés
sample_tweet = "The customer service line was terrible and delayed, absolute waste of time."
result = analyze_and_visualize_attention(sample_tweet)
print("Résultat de l'assistant de production :\n", result)
```
