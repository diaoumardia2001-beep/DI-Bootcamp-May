```python
# ==============================================================================
# PROJET COMPLET : AGENT DEVELOPPEUR GEMINI AVEC ORCHESTRATION MULTI-SERVEURS MCP
# ==============================================================================
# Environnement cible : Google Colab (1 seule cellule)
# ==============================================================================

import os
import sys
import subprocess
import shutil
import textwrap
import asyncio
from pathlib import Path

# --- ÉTAPE 1 : INSTALLATION DIRECTE DES DÉPENDANCES ---
print("⏳ 1. Installation des packages requis (cette étape peut prendre 1 à 2 minutes)...")
packages = [
    "langchain>=0.3",
    "langgraph>=0.2",
    "langchain-google-genai>=2.0",
    "google-genai>=1.0",
    "langchain-mcp-adapters==0.2.1",
    "nest_asyncio",
    "fastmcp>=2.0.0"
]

# Installation silencieuse des dépendances
subprocess.run([sys.executable, "-m", "pip", "install", "-qU"] + packages)
print("✅ Configuration des packages complétée.")

# --- ÉTAPE 2 : IMPORTS ET PATCH ASYNC ---
import nest_asyncio
from langchain_mcp_adapters.client import MultiServerMCPClient
from langchain_google_genai import ChatGoogleGenerativeAI
from langgraph.prebuilt import create_react_agent

# Application du patch d'exécution asynchrone pour l'environnement Jupyter/Colab
nest_asyncio.apply()

# --- ÉTAPE 3 : VÉRIFICATION ENVIRONNEMENT (NODE.JS) ET DOSSIER WORKSPACE ---
print("\n🔍 2. Vérification de l'environnement système...")
WORKDIR = "/content/agent_workspace"
os.makedirs(WORKDIR, exist_ok=True)
print(f"📁 Dossier de travail de l'agent configuré : {WORKDIR}")

node_exists = shutil.which("node") is not None
if not node_exists:
    print("🔄 Installation de Node.js et NPM requise...")
    subprocess.run(["apt-get", "-qq", "update"])
    subprocess.run(["apt-get", "-qq", "install", "-y", "nodejs", "npm"])
    print("✅ Node.js et NPM ont été installés avec succès.")
else:
    node_ver = os.popen('node --version').read().strip()
    print(f"✅ Node.js détecté : {node_ver}")

# --- ÉTAPE 4 : CRÉATION DU SERVEUR MCP PYTHON PERSONNALISÉ (FASTMCP) ---
print("\n💾 3. Génération du serveur MCP personnalisé (code_optimizer)...")
custom_server_path = Path("/content/custom_mcp_server.py")
custom_server_path.write_text(textwrap.dedent("""
    from fastmcp import FastMCP
    import re
    from typing import Dict, Any

    mcp = FastMCP(name="code_optimizer")

    @mcp.tool
    def audit_security(code: str) -> Dict[str, Any]:
        \"\"\"Analyses a Python script for basic security vulnerabilities (SQL injections, eval, exec, hardcoded secrets).\"\"\"
        vulnerabilities = []

        if "eval(" in code or "exec(" in code:
            vulnerabilities.append("CRITICAL: Detection of dynamic execution block ('eval' or 'exec'). This is a remote code execution risk.")

        if re.search(r"(password|passwd|secret|api_key|token)\s*=\s*['\\\"].+['\\\"]", code, re.IGNORECASE):
            vulnerabilities.append("WARNING: Possible hardcoded secret or API token detected.")

        if "execute(" in code and "%" in code:
            vulnerabilities.append("MEDIUM: Possible raw SQL injection pattern detected. Use parameterized queries instead.")

        return {
            "secure": len(vulnerabilities) == 0,
            "issues_found": len(vulnerabilities),
            "details": vulnerabilities if vulnerabilities else ["Code appears clean and compliant with basic rules."]
        }

    @mcp.tool
    def extract_todos(content: str) -> str:
        \"\"\"Extracts all TODO and FIXME comments from a code file.\"\"\"
        lines = content.split('\\n')
        todos = []
        for i, line in enumerate(lines, 1):
            if "TODO" in line or "FIXME" in line:
                todos.append(f"Line {i}: {line.strip()}")
        return "\\n".join(todos) if todos else "No TODO or FIXME tags found."

    if __name__ == "__main__":
        mcp.run(transport="stdio")
"""), encoding="utf-8")
print(f"✅ Fichier serveur enregistré sur : {custom_server_path}")

# --- ÉTAPE 5 : GESTION DE LA CLÉ API GEMINI ---
print("\n🔑 4. Configuration de l'authentification Gemini...")
if not os.getenv("GOOGLE_API_KEY"):
    try:
        from google.colab import userdata
        os.environ["GOOGLE_API_KEY"] = userdata.get("GOOGLE_API_KEY")
        print("✅ GOOGLE_API_KEY chargée depuis les secrets Google Colab.")
    except Exception:
        print("❌ Erreur : GOOGLE_API_KEY est introuvable.")
        print("👉 Veuillez ajouter votre clé API Gemini dans l'onglet 'Secrets' (icône clé 🔑 de Colab) sous le nom 'GOOGLE_API_KEY'.")
        # On définit une clé vide temporaire pour éviter un crash d'import immédiat
        os.environ["GOOGLE_API_KEY"] = "DUMMY_KEY"

# --- ÉTAPE 6 : CONFIGURATION ET CONNEXION DES SERVEURS MCP ---
mcp_connections = {
    # Serveur Filesystem standard (Node.js)
    "filesystem": {
        "transport": "stdio",
        "command": "npx",
        "args": ["-y", "@modelcontextprotocol/server-filesystem", WORKDIR],
    },
    # Serveur Fetch (Pour extraire de la doc technique sur le web)
    "fetch": {
        "transport": "stdio",
        "command": "npx",
        "args": ["-y", "@modelcontextprotocol/server-fetch"],
    },
    # Notre serveur d'optimisation de code (Python)
    "code_optimizer": {
        "transport": "stdio",
        "command": "python3",
        "args": [str(custom_server_path)],
    }
}

async def initialize_agentic_system():
    print("\n🔌 5. Initialisation du pont multi-serveurs MCP...")
    # Création du client multi-serveur (avec préfixe automatique sur les outils pour éviter les conflits)
    mcp_client = MultiServerMCPClient(mcp_connections, tool_name_prefix=True)

    # Récupération des outils traduits au format LangChain
    tools = await mcp_client.get_tools()
    print(f"✅ Connexion établie avec succès ! {len(tools)} outils MCP chargés.")

    print("\n🤖 6. Création de l'Agent intelligent Gemini-2.5-Flash...")
    # Initialisation du LLM de pointe Gemini
    llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)

    # Instructions système pour guider la logique interne du LLM
    system_instructions = f"""You are a highly competent Dev and Workspace Assistant.
    You have access to a workspace directory at '{WORKDIR}'.

    Your tools are:
    - filesystem: read, write, list files inside your workspace.
    - fetch: retrieve content from web pages to learn about library documentation.
    - code_optimizer: analyze local Python code for potential security flaws and extract TODOs.

    Guidelines:
    1. When asked to write code, write the code inside your workspace, run an audit on it using code_optimizer, and fix any security issue before finishing.
    2. If you need documentation on an external library, use the fetch tool to get details.
    3. Be transparent and present your actions step-by-step.
    """

    # Création de l'agent réactif (ReAct) autonome
    agent_executor = create_react_agent(
        model=llm,
        tools=tools,
        state_modifier=system_instructions
    )

    # --- ÉTAPE 7 : DÉROULEMENT DU SCÉNARIO DE TEST ---
    # Nous demandons une tâche complexe nécessitant l'écriture d'un fichier, son audit et sa correction.
    test_prompt = (
        "Create a new python file inside my workspace called 'db_connector.py'. "
        "Write a quick script in it to connect to a DB that uses dynamic SQL queries with variable string interpolation (using '%s' formatting). "
        "Run our security audit tool on it. "
        "If it detects any vulnerability, correct the script to use parameterized queries, save it and tell me what you did."
    )

    print(f"\n💬 Envoi de la consigne à l'agent :\n  => \"{test_prompt}\"")
    print("\n🎬 Début de la boucle de réflexion et d'exécution...")

    async for event in agent_executor.astream({"messages": [("user", test_prompt)]}):
        for node, value in event.items():
            if "messages" in value:
                last_msg = value["messages"][-1]
                # Affichage transparent des outils appelés par le LLM
                if hasattr(last_msg, 'tool_calls') and last_msg.tool_calls:
                    for tc in last_msg.tool_calls:
                        print(f"\n⚙️ [Appel Outil] '{tc['name']}' avec paramètres : {tc['args']}")
                elif last_msg.content:
                    # Affichage des réflexions et conclusions de l'agent
                    print(f"\n✍️ [{node}] : {last_msg.content[:600]}...")

# Lancement de l'orchestration globale
if os.getenv("GOOGLE_API_KEY") == "DUMMY_KEY":
    print("\n❌ L'exécution a été interrompue car la clé 'GOOGLE_API_KEY' n'est pas configurée.")
else:
    asyncio.run(initialize_agentic_system())
```
