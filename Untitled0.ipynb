### 1. Problem Statement
The primary objective of this project is to build a supervised binary classification machine learning model capable of predicting whether a loan applicant is likely to default on their loan (Target Variable: `Loan_Status` = 'N'). From a financial institution's perspective, approving a loan for an applicant who will default (False Positive) results in severe capital loss. Conversely, rejecting an applicant who would successfully repay (False Negative) results in lost interest revenue and reduced customer acquisition. The goal is to accurately classify risk to minimize credit losses while sustaining business growth.

Based on standard loan classification datasets, the following features are the most critical for identifying potential defaults, justified by risk management principles:

1. **`Credit_History` (Historical Repayment Behavior):** *Most Critical Feature.* In risk management, past behavior is statistically the strongest predictor of future performance. An applicant with a history of meeting financial obligations (indicated by a `1`) represents a significantly lower risk than someone with prior delinquencies or defaults (indicated by a `0`).
2. **`ApplicantIncome` & `CoapplicantIncome` (Financial Capacity):** Total household income dictates the gross funds available to service the debt. If the combined income is too low relative to living expenses and active liabilities, the probability of default spikes due to economic strain.
3. **`LoanAmount` relative to Income (Debt Burden Ratio):** The nominal loan value determines the required monthly installment. A large loan amount combined with low applicant income creates a highly strained payment-to-income ratio, drastically lowering the applicant's margin for financial emergencies.
4. **`Property_Area` (Collateral Security Value):** The geographical classification (Urban, Semiurban, Rural) determines the market liquidity and valuation stability of the underlying collateral. Urban and semiurban properties retain value better and can be liquidated faster by the bank to recover capital if a default occurs.

### 1. Model Selection
* **Baseline Model: Logistic Regression.** It is simple, fast to deploy, and offers high interpretability. In the banking sector, interpretability is a strict regulatory requirement; institutions must be able to explicitly explain why an applicant was denied credit.
* **Advanced Model: Extreme Gradient Boosting (XGBoost) or Random Forest.** Tree-based ensemble models excel at capturing complex, non-linear relationships (e.g., the compounding risk of a low income combined with a very long loan term). They also handle data imbalances and missing values natively.

### 2. Evaluation Strategy & Metrics
Credit risk modeling deals with highly asymmetric error costs: a False Positive (lending to a defaulting customer) is exponentially more expensive than a False Negative (rejecting a good customer).

* **Primary Optimization Metric: Recall (Sensitivity).** We want to maximize the identification of actual defaults. A high recall ensures the bank flags the maximum possible number of high-risk applicants, successfully shielding capital.
* **Secondary Metric: F1-Score / F-Beta (with $\beta=2$).** While optimizing for Recall, the bank cannot afford to reject every customer due to strict risk avoidance. The F1-Score or $F_2$-Score balances Precision and Recall to maintain commercial viability.
* **Methodology: Stratified K-Fold Cross-Validation ($K=5$).** This guarantees that each training split maintains the true, often imbalanced ratio of defaults to successful repayments, preventing optimistic or skewed performance metrics.
* **Threshold Optimization via ROC-AUC:** We will evaluate the Area Under the Receiver Operating Characteristic curve to determine the ideal probability threshold that balances capital protection against credit approval volumes.

### 1. Predicting Stock Prices ➡️ Supervised Learning (Regression)
* **Justification:** The task requires predicting a continuous numerical value (future stock price) based on historical historical inputs (past stock prices, moving averages, trading volumes, macroeconomic metrics). Because we train the model on historical input-output pairs where the target "ground truth" price is known, it is a supervised regression problem.

### 2. Organizing a Library of Books ➡️ Unsupervised Learning (Clustering)
* **Justification:** We want to group books into categories or genres based on semantic similarities (e.g., word frequencies, themes, metadata) without providing predefined category labels to the model. The algorithm (such as K-Means or Latent Dirichlet Allocation) must discover hidden structures in the unlabelled data entirely on its own, making it an unsupervised clustering problem.

### 3. Robot Navigating a Maze ➡️ Reinforcement Learning
* **Justification:** There is no static dataset or direct list of correct coordinates provided to the robot. Instead, an autonomous agent (the robot) learns to make a sequence of choices within a dynamic environment through trial and error. By receiving positive feedback/rewards for getting closer to the exit and negative feedback/penalties for hitting walls, it optimizes its decision-making policy over time.

Evaluating a machine learning solution requires an approach specific to the chosen paradigm:

#### 1. Supervised Learning (Classification Model)
* **Evaluation Strategy:** We use a confusion matrix to track exact error distributions alongside Stratified Cross-Validation. Performance is measured using Precision, Recall, F1-Score, and the ROC-AUC score.
* **Challenges & Limitations:** Real-world credit data is heavily imbalanced (the vast majority of historical loans are paid back successfully). Standard accuracy metrics can be highly misleading (e.g., a model can achieve 95% accuracy by simply approving everyone, while completely failing to flag defaults). Addressing this requires class weight balancing or synthetic sampling (SMOTE).

#### 2. Unsupervised Learning (Clustering Model)
* **Evaluation Strategy:** Lacking target labels, we must assess the mathematical structure and geometry of the generated clusters. We apply the **Elbow Method** (tracking the reduction in Within-Cluster Sum of Squares) to find the ideal number of clusters $K$, alongside the **Silhouette Score** (measuring how dense a cluster is compared to its distance from neighboring clusters, ranging from -1 to +1).
* **Challenges & Limitations:** Evaluation is inherently subjective. Because there is no objective "ground truth", a high silhouette score does not automatically mean the categories make business or practical sense. Results are highly sensitive to initial feature scaling and the selected distance metric.

#### 3. Reinforcement Learning (RL Model)
* **Evaluation Strategy:** Success is measured dynamically by tracking the **Cumulative Reward per Episode** over thousands of training runs. We look for a steady curve indicating **Convergence**, which proves that the model's action policy has stabilized. We manage the **Exploration vs. Exploitation** tradeoff by using decaying $\epsilon$-greedy strategies, forcing the agent to explore early on and exploit proven paths later.
* **Challenges & Limitations:** Designing a balanced reward function (**Reward Engineering**) is notoriously difficult; minor flaws can cause the agent to exploit loopholes instead of solving the core problem. Additionally, RL training environments can be highly unstable and demand immense computational overhead before converging safely.
