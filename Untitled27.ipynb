Step 1 : Prompt avec contrôle de ton, format et longueur

```python
prompt_v1 = (
    "Act as an internal communications specialist. Rewrite the following "
    "policy text in a friendly and clear tone for a general employee "
    "audience. Organize the information into bullet points. Paraphrase the "
    "content in your own words — do not use direct quotations from the "
    "original text. Keep the total output under 75 words.\n\n"
    "Original text: \"Employees must ensure that all remote access to "
    "internal systems is established via the approved secure VPN. Under no "
    "circumstances should unsecured connections or personal devices lacking "
    "endpoint protection be used to access proprietary data or sensitive "
    "communications.\""
)
prompt_v1
```

Step 2 : Évaluer la sortie


```python
evaluation_v1 = {
    'relevance': None,        # Est-ce fidèle au message d'origine ?
    'clarity': None,          # Plus clair pour un public non-technique ?
    'structure': None,        # Bien en bullet points ?
    'tone': None,             # Ton adapté à une comm interne ?
    'length_ok': None,        # Moins de 75 mots ?
    'factual_accuracy': None, # A-t-il inventé ou omis un détail de sécurité critique ?
    'notes': '',
}
```

Step 3 : Détecter et limiter les hallucinations


```python
prompt_v2_grounded = (
    "Act as an internal communications specialist. Rewrite the following "
    "policy text in a friendly and clear tone for a general employee "
    "audience. Organize the information into bullet points. Paraphrase the "
    "content in your own words — do not use direct quotations. Keep the "
    "total output under 75 words. "
    "Only paraphrase the content in the input text. Do not add new "
    "recommendations, technologies, or security terms that are not "
    "explicitly present in the original text.\n\n"
    "Original text: \"Employees must ensure that all remote access to "
    "internal systems is established via the approved secure VPN. Under no "
    "circumstances should unsecured connections or personal devices lacking "
    "endpoint protection be used to access proprietary data or sensitive "
    "communications.\""
)
prompt_v2_grounded
```

Step 4 : Approfondissement du paraphrasing (public stagiaire junior)


```python
prompt_v3_intern = (
    "Act as a friendly mentor explaining company policy to a new intern. "
    "Paraphrase the following policy paragraph using plain, simple language "
    "and short phrases. Use no more than 4 bullet points. Avoid any "
    "corporate or legal jargon. Keep the tone supportive and encouraging, "
    "as if helping someone new understand the rules without feeling "
    "overwhelmed.\n\n"
    "Original text: \"Employees must ensure that all remote access to "
    "internal systems is established via the approved secure VPN. Under no "
    "circumstances should unsecured connections or personal devices lacking "
    "endpoint protection be used to access proprietary data or sensitive "
    "communications.\""
)
prompt_v3_intern
```

Step 5 : Variante extraction de citation

```python
prompt_v4_quote = (
    "Read the following policy text and extract exactly one direct quote "
    "that best captures the core security requirement. Do not paraphrase — "
    "copy the sentence exactly as written in the original.\n\n"
    "Original text: \"Employees must ensure that all remote access to "
    "internal systems is established via the approved secure VPN. Under no "
    "circumstances should unsecured connections or personal devices lacking "
    "endpoint protection be used to access proprietary data or sensitive "
    "communications.\""
)
prompt_v4_quote
```

Answers to the questions:

In what kind of internal communication would quoting be more appropriate than paraphrasing?
In high-stakes compliance or legal communications — for example, a security policy reminder sent by the legal/compliance team, or an audit document where the exact wording of the rule must be preserved to avoid any ambiguity in interpretation.
When might quoting pose a risk?
When the original text is too dense, technical, or jargon-heavy for the target audience — quoting a complex legal clause word-for-word in a newsletter aimed at non-technical employees risks hurting comprehension, even though the wording is faithful. Quoting sacrifices clarity for exact precision, which isn't always the right trade-off for microlearning content.
