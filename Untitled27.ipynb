# Mini Project 1 : Meta-Analysis of Research Papers on Large Language Models (LLMs)

**Thème retenu : Rendre les LLM efficaces et accessibles — modèles fondation open-source, instruction tuning et fine-tuning sur ressources limitées**

## 1. Introduction

Les grands modèles de langage (Large Language Models, LLM) ont transformé le traitement automatique du langage naturel au cours des dernières années, en démontrant des capacités de compréhension, de génération et de raisonnement qui se rapprochent parfois du niveau humain sur des tâches variées. Cependant, l'entraînement et l'utilisation de ces modèles restent souvent réservés à des acteurs disposant d'infrastructures de calcul massives, ce qui pose un problème d'accès pour la recherche académique, les startups et les contextes à ressources limitées — une préoccupation qui rejoint directement mon propre travail sur des outils IA pour le secteur informel en Côte d'Ivoire, où le matériel disponible est souvent modeste (CPU, connectivité limitée).

L'objectif de cette méta-analyse est d'examiner comment la recherche récente a abordé le problème de l'efficacité et de l'accessibilité des LLM, à travers quatre axes complémentaires : la conception de modèles fondation ouverts et compacts, l'optimisation de l'inférence, l'instruction tuning à grande échelle, et le fine-tuning économe en mémoire. Les quatre articles sélectionnés couvrent ce thème sous des angles différents mais convergents : entraîner de meilleurs modèles avec moins de ressources, et permettre à davantage d'acteurs de les adapter à leurs besoins.

**Articles analysés :**

1. Touvron et al. (2023) — *LLaMA: Open and Efficient Foundation Language Models*, arXiv:2302.13971
2. Jiang et al. (2023) — *Mistral 7B*, arXiv:2310.06825
3. Chung et al. (2022) — *Scaling Instruction-Finetuned Language Models*, arXiv:2210.11416
4. Dettmers et al. (2023) — *QLoRA: Efficient Finetuning of Quantized LLMs*, NeurIPS 2023 / arXiv:2305.14314

## 2. Paper Summaries

### 2.1 LLaMA: Open and Efficient Foundation Language Models

**Citation complète :** Touvron, H., Lavril, T., Izacard, G., Martinet, X., Lachaux, M.-A., Lacroix, T., Rozière, B., Goyal, N., Hambro, E., Azhar, F., Rodriguez, A., Joulin, A., Grave, E., & Lample, G. (2023). *LLaMA: Open and Efficient Foundation Language Models*. arXiv:2302.13971.

**Problème de recherche.** Les modèles fondation les plus performants (GPT-3, PaLM, Chinchilla) sont propriétaires et coûteux à faire tourner en inférence. Les auteurs remettent en question l'idée selon laquelle un modèle plus grand est toujours préférable : ils s'intéressent plutôt au meilleur compromis entre budget d'entraînement et coût d'inférence.

**Solution proposée.** LLaMA est une famille de modèles fondation allant de 7 à 65 milliards de paramètres, entraînés exclusivement sur des données publiques (contrairement à des modèles propriétaires utilisant des données non partagées), sur des volumes de tokens très supérieurs à ce que suggéraient les lois d'échelle antérieures (Hoffmann et al., 2022), en privilégiant des modèles plus petits entraînés plus longtemps.

**Résultats principaux.** LLaMA-13B surpasse GPT-3 (175B) sur la plupart des benchmarks malgré une taille dix fois inférieure, et LLaMA-65B rivalise avec Chinchilla-70B et PaLM-540B. Les modèles ont été partagés avec la communauté de recherche.

**Données, architecture, évaluation.** Corpus d'entraînement composé uniquement de sources publiques (CommonCrawl, C4, Wikipedia, GitHub, livres, arXiv). Architecture Transformer décodeur classique avec des optimisations (RMSNorm, activation SwiGLU, embeddings rotatifs). Évaluation sur un large éventail de benchmarks zero-shot et few-shot (raisonnement de sens commun, question-réponse, compréhension de lecture, mathématiques, code).

### 2.2 Mistral 7B

**Citation complète :** Jiang, A. Q., Sablayrolles, A., Mensch, A., Bamford, C., Chaplot, D. S., de las Casas, D., Bressand, F., Lengyel, G., Lample, G., Saulnier, L., Lavaud, L. R., Lachaux, M.-A., Stock, P., Le Scao, T., Lavril, T., Wang, T., Lacroix, T., & El Sayed, W. (2023). *Mistral 7B*. arXiv:2310.06825.

**Problème de recherche.** Comment obtenir un modèle de taille modeste (7B) qui égale ou dépasse des modèles bien plus grands, tout en restant rapide et peu coûteux à faire tourner en inférence, y compris sur des séquences longues ?

**Solution proposée.** Mistral 7B introduit deux mécanismes d'attention optimisés : la *grouped-query attention* (GQA), qui accélère l'inférence en partageant les projections clé/valeur entre plusieurs têtes d'attention, et la *sliding window attention* (SWA), qui borne le coût de l'attention sur des séquences longues en limitant la fenêtre de contexte directement attendue à chaque couche tout en laissant l'information se propager sur des couches successives.

**Résultats principaux.** Mistral 7B dépasse Llama 2 13B sur tous les benchmarks évalués, et Llama 1 34B en raisonnement, mathématiques et génération de code. La version instruction-tunée, Mistral 7B – Instruct, dépasse Llama 2 13B – Chat dans une évaluation humaine.

**Données, architecture, évaluation.** Architecture Transformer décodeur avec GQA et SWA ; modèle et poids diffusés sous licence Apache 2.0. Évaluation sur des benchmarks de raisonnement de sens commun (HellaSwag, Winogrande, PIQA, ARC), connaissances générales, code (HumanEval, MBPP) et benchmarks agrégés (AGIEval), ainsi qu'une évaluation humaine via une plateforme dédiée pour la version Instruct.

### 2.3 Scaling Instruction-Finetuned Language Models (Flan-T5 / Flan-PaLM)

**Citation complète :** Chung, H. W., Hou, L., Longpre, S., Zoph, B., Tay, Y., Fedus, W., et al. (2022). *Scaling Instruction-Finetuned Language Models*. arXiv:2210.11416.

**Problème de recherche.** L'instruction finetuning (finetuner un modèle sur des tâches formulées comme des instructions) améliore la généralisation à des tâches non vues, mais l'effet de la mise à l'échelle du nombre de tâches, de la taille du modèle, et de l'ajout de données de raisonnement en chaîne (chain-of-thought, CoT) n'avait pas été étudié systématiquement.

**Solution proposée.** Les auteurs étendent l'instruction finetuning selon trois axes : augmenter le nombre de tâches d'entraînement jusqu'à 1 800, appliquer la méthode à des familles de modèles variées (T5, PaLM, U-PaLM) et à des tailles allant de 80 millions à 540 milliards de paramètres, et intégrer neuf jeux de données annotés en chaîne de raisonnement dans le mélange de finetuning.

**Résultats principaux.** Flan-PaLM 540B, finetuné sur 1 800 tâches, dépasse PaLM 540B de 9,4 points en moyenne et atteint 75,2 % en five-shot sur MMLU. Sans données CoT, l'instruction finetuning dégrade les performances de raisonnement en chaîne ; leur ajout permet au contraire d'améliorer toutes les évaluations. Les checkpoints Flan-T5 (dont la version small à 80M de paramètres, utilisable sur CPU) sont publiés en accès libre et atteignent de bonnes performances few-shot malgré leur petite taille, comparées à des modèles bien plus grands comme PaLM 62B.

**Données, architecture, évaluation.** Mélange de plus de 1 800 tâches issues de Muffin, T0-SF, NIV2 et de neuf jeux de données CoT annotés manuellement. Architectures encodeur-décodeur (T5) et décodeur seul (PaLM, U-PaLM). Évaluation sur MMLU, BBH (Big-Bench Hard), TyDiQA, MGSM et génération ouverte, en zero-shot, few-shot et avec CoT.

### 2.4 QLoRA: Efficient Finetuning of Quantized LLMs

**Citation complète :** Dettmers, T., Pagnoni, A., Holtzman, A., & Zettlemoyer, L. (2023). *QLoRA: Efficient Finetuning of Quantized LLMs*. Advances in Neural Information Processing Systems 36 (NeurIPS 2023) ; arXiv:2305.14314.

**Problème de recherche.** Le finetuning complet de très grands modèles (33B à 65B paramètres) nécessite des centaines de gigaoctets de mémoire GPU, hors de portée de la plupart des laboratoires ou individus.

**Solution proposée.** QLoRA combine une quantification 4-bit du modèle pré-entraîné gelé avec des adaptateurs *Low-Rank Adaptation* (LoRA) entraînables. Trois innovations techniques rendent cela possible sans perte de qualité : le format numérique NF4 (4-bit NormalFloat), optimal pour des poids suivant une distribution normale ; la double quantification, qui quantifie également les constantes de quantification pour réduire encore l'empreinte mémoire ; et des optimiseurs paginés qui gèrent les pics de mémoire pendant l'entraînement.

**Résultats principaux.** QLoRA permet de finetuner un modèle de 65 milliards de paramètres sur un unique GPU de 48 Go, tout en préservant les performances d'un finetuning complet en 16-bit. Le modèle résultant, nommé Guanaco, dépasse tous les modèles ouverts publiés précédemment sur le benchmark Vicuna, atteignant 99,3 % du niveau de performance de ChatGPT avec seulement 24 heures de finetuning sur un seul GPU.

**Données, architecture, évaluation.** Plus de 1 000 modèles finetunés pour l'étude, couvrant plusieurs familles (LLaMA, T5) et tailles (jusqu'à 65B), sur huit jeux de données d'instructions. Évaluation via le benchmark Vicuna et des comparaisons automatisées/humaines par rapport à ChatGPT.

## 3. Comparative Analysis

| Aspect | LLaMA (2023) | Mistral 7B (2023) | Flan-T5 / Flan-PaLM (2022) | QLoRA (2023) |
|---|---|---|---|---|
| **Objectif principal** | Modèles fondation compétitifs entraînés sur données publiques | Modèle compact optimisé pour l'inférence rapide | Généralisation via instruction finetuning à grande échelle | Finetuning mémoire-efficace de très grands modèles |
| **Domaine de problème** | Pré-entraînement / architecture | Pré-entraînement / architecture d'attention | Finetuning supervisé multi-tâches | Finetuning paramétrique (PEFT) |
| **Taille des modèles** | 7B à 65B | 7B | 80M à 540B | Jusqu'à 65B (finetuning) |
| **Innovation clé** | Ratio données/paramètres plus élevé que les lois d'échelle classiques | Grouped-query attention + sliding window attention | Mélange massif de tâches + données chain-of-thought | Quantification NF4 + double quantification + optimiseurs paginés |
| **Stratégie d'entraînement** | Pré-entraînement complet sur données publiques | Pré-entraînement complet, poids ouverts | Finetuning supervisé sur instructions | Finetuning LoRA sur backbone quantifié 4-bit |
| **Benchmarks principaux** | HellaSwag, TriviaQA, MMLU, benchmarks de code | HellaSwag, ARC, HumanEval, MBPP, AGIEval | MMLU, BBH, TyDiQA, MGSM | Vicuna benchmark, comparaison à ChatGPT |
| **Reproductibilité** | Poids diffusés à la communauté académique (accès contrôlé à l'origine) | Poids et code sous licence Apache 2.0, totalement ouverts | Checkpoints Flan-T5 publiés en accès libre | Code et méthode publiés, checkpoints Guanaco publiés |
| **Limite principale** | Toujours coûteux à pré-entraîner from scratch | Efficace en inférence mais nécessite encore du fine-tuning pour usages spécialisés | Le CoT est nécessaire pour ne pas dégrader le raisonnement | Le gain mémoire ne réduit pas le coût de l'inférence finale (le modèle de base reste grand) |

**Points de comparaison qualitatifs :**

- **Architectures et innovations** : LLaMA et Mistral partagent une architecture Transformer décodeur, mais Mistral pousse plus loin l'optimisation de l'attention (GQA, SWA) spécifiquement pour réduire le coût d'inférence à taille de modèle fixe, alors que LLaMA optimise surtout le ratio données/paramètres au moment du pré-entraînement.
- **Approche de l'efficacité** : les quatre papiers abordent l'efficacité sous des angles différents — LLaMA vise l'efficacité au pré-entraînement, Mistral l'efficacité à l'inférence, Flan-T5/Flan-PaLM l'efficacité en généralisation (moins de données nécessaires par tâche grâce à l'instruction tuning), et QLoRA l'efficacité mémoire au moment du finetuning.
- **Complémentarité** : ces approches ne sont pas concurrentes mais cumulables. Dans la pratique (et c'est ce que j'observe dans mes propres projets avec des modèles comme flan-t5-small), on peut partir d'un modèle fondation efficace type LLaMA/Mistral, lui appliquer un instruction tuning façon Flan, puis le spécialiser sur des données propres via QLoRA — trois briques indépendantes qui s'assemblent.

## 4. Insights and Reflection

**Tendances communes.** Les quatre articles illustrent un mouvement de fond dans la recherche sur les LLM depuis 2022 : l'attention se déplace progressivement de « comment construire le plus grand modèle possible » vers « comment obtenir la meilleure performance pour un budget de calcul, de mémoire ou de données donné ». Que ce soit au niveau du pré-entraînement (LLaMA), de l'architecture d'inférence (Mistral), du finetuning supervisé (Flan) ou du finetuning paramétrique (QLoRA), chaque papier remet en question une hypothèse implicite de la génération précédente de modèles géants propriétaires.

**Approches les plus prometteuses.** L'instruction finetuning à grande échelle avec données CoT (Flan) et la quantification combinée à LoRA (QLoRA) semblent particulièrement porteuses pour des contextes à ressources limitées comme le mien : elles permettent d'obtenir des modèles utiles sans disposer d'un cluster de GPU. La combinaison GQA + SWA de Mistral est également très pertinente pour des applications nécessitant une latence faible, comme des assistants vocaux en langue locale (Dioula, Baoulé) où le temps de réponse compte autant que la qualité.

**Limites communément reconnues.** Tous les papiers reconnaissent, explicitement ou implicitement, un compromis entre efficacité et exhaustivité : les petits modèles (Flan-T5-small, ou même Mistral 7B face à des modèles de plusieurs centaines de milliards de paramètres) restent en retrait sur des tâches de raisonnement complexe ou de connaissances factuelles très spécifiques. QLoRA, de son côté, réduit le coût du finetuning mais pas celui de l'inférence : un modèle de 65B reste un modèle de 65B à faire tourner, même s'il a été finetuné à moindre coût.

**Directions futures.** Les quatre travaux ouvrent la voie à des architectures encore plus légères combinées à des méthodes de finetuning paramétrique, à une meilleure compression pour le déploiement sur appareils mobiles ou edge, et à des jeux d'instructions plus représentatifs de langues et contextes sous-représentés — un axe qui rejoint directement mon intérêt pour des LLM capables de fonctionner dans des environnements à faible connectivité et en langues locales ouest-africaines.

## 5. Conclusion

Cette méta-analyse a examiné quatre contributions majeures à la recherche sur l'efficacité des grands modèles de langage : LLaMA, qui repense le compromis données/paramètres au pré-entraînement ; Mistral 7B, qui optimise l'architecture d'attention pour l'inférence ; Flan-T5/Flan-PaLM, qui démontre la puissance de l'instruction finetuning à grande échelle avec raisonnement en chaîne ; et QLoRA, qui rend le finetuning de très grands modèles accessible sur un seul GPU grand public.

Le fil conducteur qui relie ces travaux est clair : la performance brute d'un modèle ne suffit plus à évaluer sa valeur pratique ; le coût d'entraînement, le coût d'inférence et l'accessibilité des poids et du code deviennent des critères tout aussi centraux. Cette évolution démocratise progressivement l'accès aux LLM pour des acteurs disposant de ressources limitées — chercheurs indépendants, startups, ou porteurs de projets dans des contextes comme l'Afrique de l'Ouest, où l'infrastructure de calcul reste un facteur limitant majeur. Le champ semble se diriger vers une convergence de ces techniques (modèles fondation efficaces + instruction tuning + finetuning paramétrique), plutôt que vers une simple course à la taille des modèles.
