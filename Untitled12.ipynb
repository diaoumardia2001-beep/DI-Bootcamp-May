Step 1: Dependencies and Setup
First, let's pull in your deep learning toolkit. We will set deterministic seeds to guarantee reproducibility every time you run the workbook.

```python
import os
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import r2_score

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

# Set seed for reproducibility
def set_seeds(seed=42):
    random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.backends.cudnn.deterministic = True

set_seeds()

# Check for GPU
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f" Using execution device: {device}")
```

Step 2 & 3: Synthetic Data Generation & Preprocessing Pipeline
Since stock datasets vary, this step constructs a synthetic stock price dataframe with classic time-series movements (Close, Open, High, Low, Volume). It maps a sliding lookback window of 30 days (WINDOW_SIZE) to project the next day's price (HORIZON = 1).

```python
# 1. Simulate mock Stock market historical pricing dataframe
print(" Simulating financial dataset...")
dates = pd.date_range(start="2020-01-01", periods=1500, freq='D')
np.random.seed(42)
close_prices = 100 + np.cumsum(np.random.normal(0.1, 1.2, size=1500))

df = pd.DataFrame({
    'Open': close_prices - np.random.uniform(0.5, 2.0, 1500),
    'High': close_prices + np.random.uniform(0.5, 2.0, 1500),
    'Low': close_prices - np.random.uniform(1.0, 3.0, 1500),
    'Close': close_prices,
    'Volume': np.random.randint(10000, 500000, size=1500)
}, index=dates)

# 2. Create target column: Next Day's Closing Price
df['Target'] = df['Close'].shift(-1)
df = df.dropna()  # Remove last row because it doesn't have a future target step

# 3. Chronological splits (Train: 70%, Val: 15%, Test: 15%)
n = len(df)
train_idx = int(n * 0.7)
val_idx = int(n * 0.85)

train_df = df.iloc[:train_idx]
val_df = df.iloc[train_idx:val_idx]
test_df = df.iloc[val_idx:]

# 4. Fit scaling operators cleanly using only the training slice
feature_cols = ['Open', 'High', 'Low', 'Close', 'Volume']
feature_scaler = MinMaxScaler(feature_range=(0, 1))
target_scaler = MinMaxScaler(feature_range=(0, 1))

train_feats_scaled = feature_scaler.fit_transform(train_df[feature_cols].values)
train_targ_scaled = target_scaler.fit_transform(train_df[['Target']].values)

val_feats_scaled = feature_scaler.transform(val_df[feature_cols].values)
val_targ_scaled = target_scaler.transform(val_df[['Target']].values)

test_feats_scaled = feature_scaler.transform(test_df[feature_cols].values)
test_targ_scaled = target_scaler.transform(test_df[['Target']].values)

# --- Custom PyTorch Dataset Constructor ---
class StockDataset(Dataset):
    def __init__(self, features, targets, window_size=30):
        self.features = torch.tensor(features, dtype=torch.float32)
        self.targets = torch.tensor(targets, dtype=torch.float32)
        self.window_size = window_size

    def __len__(self):
        return len(self.features) - self.window_size

    def __getitem__(self, idx):
        x = self.features[idx : idx + self.window_size]
        y = self.targets[idx + self.window_size]
        return x, y

WINDOW_SIZE = 30
train_dataset = StockDataset(train_feats_scaled, train_targ_scaled, WINDOW_SIZE)
val_dataset = StockDataset(val_feats_scaled, val_targ_scaled, WINDOW_SIZE)
test_dataset = StockDataset(test_feats_scaled, test_targ_scaled, WINDOW_SIZE)

# Combine into PyTorch Dataloaders for streamlined batching
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)

print(f"Data ready. Training batches: {len(train_loader)} | Validation batches: {len(val_loader)}")
```

Step 4: Define the PyTorch LSTM Architecture
Here, we define our custom recurrent model using nn.Module. Note that while your assignment guidelines mention LSTM, it also hints at GRU; this structure uses an LSTM stack as the structural base, layered with a dropout regularization head to stop overfitting.

```python
class StockLSTM(nn.Module):
    def __init__(self, input_dim, hidden_dim, num_layers=2, dropout_prob=0.2):
        super(StockLSTM, self).__init__()
        self.hidden_dim = hidden_dim
        self.num_layers = num_layers

        # LSTM Layer: batch_first=True maps inputs as [Batch Size, Seq Length, Features]
        self.lstm = nn.LSTM(
            input_size=input_dim,
            hidden_size=hidden_dim,
            num_layers=num_layers,
            batch_first=True,
            dropout=dropout_prob if num_layers > 1 else 0.0
        )

        self.dropout = nn.Dropout(dropout_prob)
        self.fc = nn.Linear(hidden_dim, 1) # Single linear prediction output

    def forward(self, x):
        # Initialize initial structural internal cell states with zeroes
        h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_dim).to(x.device)
        c0 = torch.zeros(self.num_layers, x.size(0), self.hidden_dim).to(x.device)

        # Propagate forward through recurrent cell units
        out, (hn, cn) = self.lstm(x, (h0, c0))

        # Extract the hidden state slice of the final sequence time step
        out = out[:, -1, :]
        out = self.dropout(out)
        out = self.fc(out)
        return out

# Instantiate model
model = StockLSTM(input_dim=len(feature_cols), hidden_dim=64, num_layers=2).to(device)
print(model)
```

Step 5: Training and Validation Loops
Unlike Keras, PyTorch requires explicit structural backpropagation manual updates. We set up our optimization matrix using Adam optimization tracking across MSE Loss.

```python
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

epochs = 20
train_losses, val_losses = [], []

print(" Starting training process...")
for epoch in range(epochs):
    # --- TRAINING PHASE ---
    model.train()
    running_train_loss = 0.0
    for batch_x, batch_y in train_loader:
        batch_x, batch_y = batch_x.to(device), batch_y.to(device)

        optimizer.zero_grad()          # Reset cached historical gradients
        predictions = model(batch_x)    # Forward pass execution
        loss = criterion(predictions, batch_y)
        loss.backward()                 # Backpropagation
        optimizer.step()                # Weights optimization

        running_train_loss += loss.item() * batch_x.size(0)

    epoch_train_loss = running_train_loss / len(train_loader.dataset)
    train_losses.append(epoch_train_loss)

    # --- VALIDATION PHASE ---
    model.eval()
    running_val_loss = 0.0
    with torch.no_grad(): # Disable gradient tracking to conserve execution memory
        for batch_x, batch_y in val_loader:
            batch_x, batch_y = batch_x.to(device), batch_y.to(device)
            predictions = model(batch_x)
            loss = criterion(predictions, batch_y)
            running_val_loss += loss.item() * batch_x.size(0)

    epoch_val_loss = running_val_loss / len(val_loader.dataset)
    val_losses.append(epoch_val_loss)

    if (epoch + 1) % 2 == 0 or epoch == 0:
        print(f"Epoch {epoch+1:02d}/{epochs} | Train Loss: {epoch_train_loss:.6f} | Val Loss: {epoch_val_loss:.6f}")

# Plot loss convergence history profile
plt.figure(figsize=(7, 3.5))
plt.plot(train_losses, label='Train Loss', color='royalblue')
plt.plot(val_losses, label='Val Loss', color='crimson')
plt.title('PyTorch System Convergence Curves')
plt.xlabel('Epochs')
plt.ylabel('MSE Loss Scale')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

Step 6: Evaluation and $R^2$ ScoringTo evaluate our model, we run inference on our testing pipeline, invert our MinMax scalar transformation to restore real dollar values, and compute the final $R^2$ evaluation accuracy score.

```python
model.eval()
raw_predictions, raw_targets = [], []

with torch.no_grad():
    for batch_x, batch_y in test_loader:
        batch_x = batch_x.to(device)
        preds = model(batch_x)

        raw_predictions.append(preds.cpu().numpy())
        raw_targets.append(batch_y.numpy())

# Flatten array footprints
all_preds = np.vstack(raw_predictions)
all_targets = np.vstack(raw_targets)

# Invert processing transforms to restore native currency metrics
true_prices = target_scaler.inverse_transform(all_targets)
predicted_prices = target_scaler.inverse_transform(all_preds)

# Calculate goodness of fit performance score (R2)
r2 = r2_score(true_prices, predicted_prices)

print("\n==========================================")
print(" EVALUATION RESULTS SUMMARY")
print("==========================================")
print(f"R² Score Accuracy Level: {r2:.4f}")
if r2 > 0.80:
    print(" High predictive performance achieved!")
else:
    print(" Model represents standard baseline behaviors.")

# Save down training checkpoint states for production deployment
torch.save(model.state_dict(), 'stock_lstm_model.pth')
print("\n Model artifact successfully saved to 'stock_lstm_model.pth'")
```
