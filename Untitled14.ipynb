```python
# Prefilled. Just copy and execute.
import os, math, re, random
from glob import glob
from pathlib import Path
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

np.random.seed(42); tf.random.set_seed(42)

# Paths - change if needed
DATA_ROOT = Path("data/cats_dogs")
train_dir = (DATA_ROOT / "train" / "train") if (DATA_ROOT / "train" / "train").exists() else (DATA_ROOT / "train")
test_dir  = (DATA_ROOT / "test"  / "test")  if (DATA_ROOT / "test"  / "test").exists()  else (DATA_ROOT / "test")

IMG_HEIGHT, IMG_WIDTH = 180, 180
batch_size = 32
seed = 1337

# Build DataFrames from folders
def build_df_from_folder(folder: Path, labeled: bool=True):
    exts = ('*.jpg','*.jpeg','*.png','*.bmp')
    files = []
    for ex in exts:
        files.extend(glob(str(folder / '**' / ex), recursive=True))
    if not files:
        raise FileNotFoundError(f"No images found under {folder}")
    rows = []
    for f in files:
        if labeled:
            name = Path(f).name.lower()
            parent = Path(f).parent.name.lower()
            if parent in {"cat","cats"}:
                label = "cat"
            elif parent in {"dog","dogs"}:
                label = "dog"
            else:
                if re.search(r'(^|[^a-z])cat([^a-z]|$)', name): label = "cat"
                elif re.search(r'(^|[^a-z])dog([^a-z]|$)', name): label = "dog"
                else:
                    continue
            rows.append({"filepath": f, "label": label})
        else:
            rows.append({"filepath": f})
    return pd.DataFrame(rows)

df_train_full = build_df_from_folder(train_dir, labeled=True)
df_test_full  = build_df_from_folder(test_dir,  labeled=False)

# Train validation split
from sklearn.model_selection import train_test_split
df_tr, df_val = train_test_split(
    df_train_full, test_size=0.2, stratify=df_train_full["label"], random_state=seed
)

# Generators
train_gen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=45,
    width_shift_range=0.15,
    height_shift_range=0.15,
    zoom_range=0.5,
    horizontal_flip=True,
)
val_gen = ImageDataGenerator(rescale=1./255)
test_gen = ImageDataGenerator(rescale=1./255)

train_flow = train_gen.flow_from_dataframe(
    df_tr, x_col="filepath", y_col="label",
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    class_mode="binary", batch_size=batch_size,
    shuffle=True, seed=seed, validate_filenames=False
)
val_flow = val_gen.flow_from_dataframe(
    df_val, x_col="filepath", y_col="label",
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    class_mode="binary", batch_size=batch_size,
    shuffle=False, validate_filenames=False
)
# Unlabeled test for inference only
test_flow = test_gen.flow_from_dataframe(
    df_test_full, x_col="filepath", y_col=None,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    class_mode=None, batch_size=batch_size,
    shuffle=False, validate_filenames=False
)

print({"train": train_flow.samples, "val": val_flow.samples, "test": test_flow.samples,
       "class_indices": train_flow.class_indices})
```

3 & 4. Model Architecture and Optimization SetupModel Architecture and Optimization DescriptionThe planned Convolutional Neural Network (CNN) is structured around three successive convolutional blocks. The first block utilizes 32 filters of size $3 \times 3$ to capture low-level edges and baseline textures. The second block employs 64 filters to assemble these primitive shapes, while the third block uses 128 filters to extract high-level complex patterns (such as eyes and body silhouettes). Each convolutional block is immediately followed by a $2 \times 2$ MaxPooling2D layer, which halves the spatial dimensions. This provides the model with translational invariance while significantly reducing the parameter footprint to prevent overfitting.Following the feature extraction layers, a Flatten layer flattens the multidimensional feature maps into a single vector, which is then fed into a dense hidden layer with 128 neurons. For regularization, a Dropout layer with a rate of 0.5 is integrated right after this dense layer. It randomly deactivates half of the neurons during each training step, forcing the network to learn robust, co-independent, and redundant internal representations. The final classification step is handled by a dense output layer containing a single Sigmoid activation unit. This setup is mathematically ideal for modeling a Bernoulli target distribution on a raw probability scale within $[0, 1]$, and is optimized using the Binary Cross-Entropy loss function.Optimization Setup Justification: We select the Adam optimizer with an initial learning rate of 0.001, which serves as a highly robust industry standard. It ensures rapid convergence dynamics by employing adaptive first and second-order moments tracking. A batch size of 32 is chosen to provide an optimal compromise between stable gradient trajectory updates and GPU/CPU memory efficiency.

```python
from tensorflow.keras import models, layers, optimizers
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

def build_model():
    return models.Sequential([
        layers.Input(shape=(IMG_HEIGHT, IMG_WIDTH, 3)),
        layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
        layers.MaxPooling2D((2, 2)),
        layers.Flatten(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(1, activation='sigmoid')
    ])

model_augmented = build_model()
model_augmented.compile(optimizer=optimizers.Adam(learning_rate=0.001), loss='binary_crossentropy', metrics=['accuracy'])

# Configuration des outils de régulation dynamique
callbacks_list = [
    EarlyStopping(monitor='val_loss', patience=4, restore_best_weights=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, verbose=1)
]

print(" Entraînement du modèle augmenté...")
history_augmented = model_augmented.fit(
    train_flow, validation_data=val_flow, epochs=15, callbacks=callbacks_list, verbose=1
)
```

Overfitting Detection & Mitigation
How Overfitting is Detected: Overfitting is traditionally identified when the training loss continuously decreases while the validation loss plateaus or begins to diverge upward. In this specific run, classic overfitting has not fully set in yet; instead, the model is experiencing an early training bottleneck where it struggles to generalise past a baseline threshold due to the aggressive data augmentation applied relative to the small size of the dataset (1,600 training images).

Mitigation Applied: The integration of the regularisation pipeline successfully protected the workflow. The Dropout(0.5) layer prevented the dense layers from memorising the raw pixels of the limited training set. Simultaneously, the EarlyStopping callback acted as a critical safety net, interrupting the process the moment the validation performance degraded, ensuring that the final model saved is the most generalised version (from Epoch 3) rather than an overfitted one.

9. Handling Class Imbalances
Class Imbalance Analysis
As demonstrated in Section 2, the extracted dataset exhibits a perfect equilibrium of 50% cats and 50% dogs for the training set. Therefore, this To-Do item is not applicable to our specific case. >
If a severe class imbalance had been present (e.g., 90% cats and 10% dogs), the model would likely lean towards predicting the majority class to maximize overall accuracy, thereby ignoring the minority class features. In such a scenario, calculating class weights that are inversely proportional to class frequencies via the class_weight parameter during training becomes crucial. This mathematical adjustment penalizes errors committed on the minority class much more heavily, forcing the network to pay equal attention to both distributions and significantly boosting the recall score for the underrepresented class.

11. Extensions (Transfer Learning with MobileNetV2)
Justification of Expected Benefit:

Leveraging MobileNetV2 allows the pipeline to benefit from rich, pre-trained low-level feature extractors (such as fur textures, edges, and complex shapes) optimized on millions of images from the ImageNet dataset. Instead of learning spatial hierarchies from scratch—which is highly data-inefficient—the model only needs to map these robust existing features to our specific target classes. Consequently, this architecture typically achieves an accuracy exceeding 90% within just a few epochs, while drastically slashing total training time and computational overhead compared to training a baseline CNN from scratch.
