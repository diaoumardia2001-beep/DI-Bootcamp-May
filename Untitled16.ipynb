```python
import math
import torch
import torch.nn as nn

class SingleHeadAttention(nn.Module):
    def __init__(self, hidden_dim):
        super().__init__()
        self.hidden_dim = hidden_dim
        self.d_k = hidden_dim

        # Projections linéaires pour Query, Key, et Value
        self.q_proj = nn.Linear(hidden_dim, hidden_dim)
        self.k_proj = nn.Linear(hidden_dim, hidden_dim)
        self.v_proj = nn.Linear(hidden_dim, hidden_dim)

        self.softmax = nn.Softmax(dim=-1)

    def forward(self, x):
        # x shape: (batch_size, seq_len, hidden_dim)
        batch_size, seq_len, _ = x.size()

        # 1. Projection linéaire
        Q = self.q_proj(x)  # (B, S, D)
        K = self.k_proj(x)  # (B, S, D)
        V = self.v_proj(x)  # (B, S, D)

        # 2. Calcul des scores (Produit scalaire scalé)
        # (B, S, D) x (B, D, S) -> (B, S, S)
        scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(self.d_k)

        # 3. Application du Softmax pour obtenir la carte d'attention
        attention_weights = self.softmax(scores)

        # 4. Mélange des valeurs : (B, S, S) x (B, S, D) -> (B, S, D)
        output = torch.matmul(attention_weights, V)

        return output, attention_weights

# --- Validation des Shapes ---
if __name__ == "__main__":
    B, S, D = 4, 12, 64  # Batch, Seq_len, Hidden_dim
    dummy_input = torch.randn(B, S, D)

    sha = SingleHeadAttention(hidden_dim=D)
    out, weights = sha(dummy_input)

    print("=== Validation Single-Head Attention ===")
    print(f"Shape Entrée   : {dummy_input.shape}")
    print(f"Shape Sortie   : {out.shape} (Doit être identique à l'entrée)")
    print(f"Shape Poids Attn: {weights.shape} (Batch, Seq_len, Seq_len)")
```

```python
class MultiHeadAttention(nn.Module):
    def __init__(self, hidden_dim, num_heads, dropout=0.1):
        super().__init__()
        assert hidden_dim % num_heads == 0, "hidden_dim doit être divisible par num_heads"

        self.hidden_dim = hidden_dim
        self.num_heads = num_heads
        self.head_dim = hidden_dim // num_heads

        # Une seule grande couche linéaire pour Q, K, V pour optimiser les calculs
        self.qkv_proj = nn.Linear(hidden_dim, hidden_dim * 3)
        self.out_proj = nn.Linear(hidden_dim, hidden_dim)

        self.dropout = nn.Dropout(dropout)
        self.softmax = nn.Softmax(dim=-1)

    def forward(self, x):
        batch_size, seq_len, hidden_dim = x.size()

        # 1. Projection et séparation en Q, K, V
        qkv = self.qkv_proj(x)  # (B, S, 3 * D)
        qkv = qkv.reshape(batch_size, seq_len, 3, self.num_heads, self.head_dim)
        qkv = qkv.permute(2, 0, 3, 1, 4)  # (3, B, H, S, head_dim)

        Q, K, V = qkv[0], qkv[1], qkv[2]  # Chacun a la forme : (B, H, S, head_dim)

        # 2. Calcul de l'attention par tête
        scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(self.head_dim)  # (B, H, S, S)
        attn_weights = self.softmax(scores)
        attn_weights = self.dropout(attn_weights)

        # 3. Fusion et concaténation des têtes
        context = torch.matmul(attn_weights, V)  # (B, H, S, head_dim)
        context = context.permute(0, 2, 1, 3).reshape(batch_size, seq_len, hidden_dim)  # (B, S, D)

        # 4. Projection de sortie finale
        output = self.out_proj(context)
        return output, attn_weights

# --- Validation des Shapes ---
if __name__ == "__main__":
    mha = MultiHeadAttention(hidden_dim=64, num_heads=8)
    out_mha, weights_mha = mha(dummy_input)
    print("\n=== Validation Multi-Head Attention ===")
    print(f"Shape Sortie   : {out_mha.shape}")
    print(f"Shape Poids Attn: {weights_mha.shape} (Batch, Heads, Seq_len, Seq_len)")
```

```python
class TransformerEncoderBlock(nn.Module):
    def __init__(self, hidden_dim, num_heads, ffn_dim, dropout=0.1):
        super().__init__()
        self.mha = MultiHeadAttention(hidden_dim, num_heads, dropout)
        self.norm1 = nn.LayerNorm(hidden_dim)
        self.norm2 = nn.LayerNorm(hidden_dim)

        self.ffn = nn.Sequential(
            nn.Linear(hidden_dim, ffn_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(ffn_dim, hidden_dim)
        )
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # Sous-couche 1 : Attention + Connexion Résiduelle + Norm
        attn_out, weights = self.mha(x)
        x = self.norm1(x + self.dropout(attn_out))

        # Sous-couche 2 : Feed-Forward + Connexion Résiduelle + Norm
        ffn_out = self.ffn(x)
        x = self.norm2(x + self.dropout(ffn_out))
        return x, weights

# --- Exemple de Classificateur pour l'NLI (Premise + Hypothesis) ---
class CustomNLIClassifier(nn.Module):
    def __init__(self, vocab_size, hidden_dim, num_heads, num_classes=3):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, hidden_dim)
        self.encoder = TransformerEncoderBlock(hidden_dim, num_heads, ffn_dim=hidden_dim*4)
        self.classifier = nn.Linear(hidden_dim, num_classes)

    def forward(self, x):
        x = self.embedding(x)
        encoded, weights = self.encoder(x)
        # Pooling moyen global sur la séquence
        pooled = encoded.mean(dim=1)
        logits = self.classifier(pooled)
        return logits, weights
```

Réflexion et Analyse Comparative1. Différences conceptuelles d'attentionSingle-Head Attention : Force le modèle à se concentrer sur une seule relation à la fois. Si un mot a une relation à la fois syntaxique (trouver son verbe) et sémantique (résoudre un pronom), une seule tête risque d'écraser l'une des deux informations.Multi-Head Attention : Divise l'espace en plusieurs "sous-espaces de projection". Cela permet à la tête 1 de suivre la grammaire, à la tête 2 de repérer les sujets, et à la tête 3 de l'associer aux adjectifs de manière totalement parallèle.Cross-Attention : Contrairement à la Self-Attention (où $Q$, $K$, $V$ viennent de la même phrase), la Cross-Attention prend ses requêtes ($Q$) d'une séquence cible (comme le décodeur d'une traduction) et ses clés/valeurs ($K$, $V$) d'une séquence source (l'encodeur). Cela permet de lier deux langues ou deux modalités différentes (ex: texte et image).2. Conclusion sur le comportement de l'attentionLes Transformers ne possèdent pas de biais inductif pour le langage (ils ne savent pas nativement qu'un mot suit un autre). C'est pour cela que sans encodage positionnel (Positional Encoding) et sans un volume colossal de données, notre modèle custom aura du mal à égaler un modèle comme DistilBERT.L'analyse des cartes d'attention (attention maps) montre que les modèles pré-entraînés ont appris à structurer leur focus sur des éléments clés (comme le token [CLS] ou la ponctuation), là où un modèle entraîné sur un petit dataset à partir de zéro aura une carte d'attention beaucoup plus diffuse et confuse.
