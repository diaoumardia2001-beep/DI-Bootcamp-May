```python
import math
import torch
import torch.nn as nn

class SingleHeadAttention(nn.Module):
    def __init__(self, hidden_dim):
        super().__init__()
        self.hidden_dim = hidden_dim
        self.d_k = hidden_dim

        # Linear projections for Query, Key, and Value
        self.q_proj = nn.Linear(hidden_dim, hidden_dim)
        self.k_proj = nn.Linear(hidden_dim, hidden_dim)
        self.v_proj = nn.Linear(hidden_dim, hidden_dim)

        self.softmax = nn.Softmax(dim=-1)

    def forward(self, x):
        # Input shape: (batch_size, seq_len, hidden_dim)
        batch_size, seq_len, _ = x.size()

        # Project inputs
        Q = self.q_proj(x) # (B, S, D)
        K = self.k_proj(x) # (B, S, D)
        V = self.v_proj(x) # (B, S, D)

        # Calculate scaled dot-product scores
        # (B, S, D) x (B, D, S) -> (B, S, S)
        scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(self.d_k)

        # Softmax normalized distribution (Attention Map)
        attention_weights = self.softmax(scores)

        # Mix values: (B, S, S) x (B, S, D) -> (B, S, D)
        output = torch.matmul(attention_weights, V)

        return output, attention_weights

# --- Shape Validation & Testing ---
if __name__ == "__main__":
    batch_size = 4
    seq_len = 12
    hidden_dim = 64

    dummy_input = torch.randn(batch_size, seq_len, hidden_dim)
    sha_module = SingleHeadAttention(hidden_dim=hidden_dim)

    out, weights = sha_module(dummy_input)
    print("=== Single-Head Attention Validation ===")
    print(f"Input Shape:     {dummy_input.shape}")
    print(f"Output Shape:    {out.shape}")
    print(f"Weights Shape:   {weights.shape} (Batch, Seq_Len, Seq_Len)")
```

```python
class MultiHeadAttention(nn.Module):
    def __init__(self, hidden_dim, num_heads, dropout=0.1):
        super().__init__()
        assert hidden_dim % num_heads == 0, "hidden_dim must be divisible by num_heads"

        self.hidden_dim = hidden_dim
        self.num_heads = num_heads
        self.head_dim = hidden_dim // num_heads

        # Combined projection layer for efficiency
        self.qkv_proj = nn.Linear(hidden_dim, hidden_dim * 3)
        self.out_proj = nn.Linear(hidden_dim, hidden_dim)

        self.dropout = nn.Dropout(dropout)
        self.softmax = nn.Softmax(dim=-1)

    def forward(self, x):
        batch_size, seq_len, hidden_dim = x.size()

        # 1. Project and split into Q, K, V
        # Shape: (B, S, 3 * D)
        qkv = self.qkv_proj(x)

        # Reshape to separate Q, K, V and slice
        # Shape: (B, S, 3, H, head_dim)
        qkv = qkv.reshape(batch_size, seq_len, 3, self.num_heads, self.head_dim)
        # Permute to make it: (3, B, H, S, head_dim)
        qkv = qkv.permute(2, 0, 3, 1, 4)
        Q, K, V = qkv[0], qkv[1], qkv[2] # Each is (B, H, S, head_dim)

        # 2. Scaled Dot-Product attention across sub-spaces
        # (B, H, S, head_dim) x (B, H, head_dim, S) -> (B, H, S, S)
        scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(self.head_dim)
        attn_weights = self.softmax(scores)
        attn_weights = self.dropout(attn_weights)

        # 3. Value Mixing & Head Concatenation
        # (B, H, S, S) x (B, H, S, head_dim) -> (B, H, S, head_dim)
        context = torch.matmul(attn_weights, V)

        # Transpose and reshape back to original dimension: (B, S, D)
        context = context.permute(0, 2, 1, 3).reshape(batch_size, seq_len, hidden_dim)

        # 4. Final linear output projection
        output = self.out_proj(context)
        return output, attn_weights

# --- Shape Validation ---
if __name__ == "__main__":
    mha_module = MultiHeadAttention(hidden_dim=64, num_heads=8)
    out, weights = mha_module(dummy_input)
    print("\n=== Multi-Head Attention Validation ===")
    print(f"Output Shape:    {out.shape} (Must match Input)")
    print(f"Weights Shape:   {weights.shape} (Batch, Heads, Seq_Len, Seq_Len)")
```

```python
class TransformerEncoderBlock(nn.Module):
    def __init__(self, hidden_dim, num_heads, ffn_dim, dropout=0.1):
        super().__init__()
        self.mha = MultiHeadAttention(hidden_dim, num_heads, dropout)
        self.norm1 = nn.LayerNorm(hidden_dim)
        self.norm2 = nn.LayerNorm(hidden_dim)

        self.ffn = nn.Sequential(
            nn.Linear(hidden_dim, ffn_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(ffn_dim, hidden_dim)
        )
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # Block 1: Attention + Sublayer Connection
        attn_out, weights = self.mha(x)
        x = self.norm1(x + self.dropout(attn_out))

        # Block 2: FFN + Sublayer Connection
        ffn_out = self.ffn(x)
        x = self.norm2(x + self.dropout(ffn_out))
        return x, weights

class CustomNLIClassifier(nn.Module):
    def __init__(self, vocab_size, hidden_dim, num_heads, num_classes=3):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, hidden_dim)
        self.encoder = TransformerEncoderBlock(hidden_dim, num_heads, ffn_dim=hidden_dim*4)
        self.classifier = nn.Linear(hidden_dim, num_classes)

    def forward(self, x):
        # x: Token tensors combining [Premise + Hypothesis]
        x = self.embedding(x)
        encoded, weights = self.encoder(x)

        # Global average pooling over token sequence
        pooled = encoded.mean(dim=1)
        logits = self.classifier(pooled)
        return logits, weights

# --- Dummy Execution Loop Simulation ---
if __name__ == "__main__":
    vocab_size = 1000
    model = CustomNLIClassifier(vocab_size=vocab_size, hidden_dim=64, num_heads=4)
    optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3)
    criterion = nn.CrossEntropyLoss()

    # Simulating data loading of a sequence pair batch (Batch_size=2, Seq_len=10)
    mock_batch = torch.randint(0, vocab_size, (2, 10))
    mock_labels = torch.tensor([0, 2]) # entailment, contradiction, neutral

    # Train Step Mock
    model.train()
    optimizer.zero_grad()
    logits, weights = model(mock_batch)
    loss = criterion(logits, mock_labels)
    loss.backward()
    optimizer.step()

    print("\n=== Custom Encoder Pipeline Check ===")
    print(f"Loss step validated: {loss.item():.4f}")
    print(f"Logits shape:        {logits.shape}")
```

Comprehensive Paradigms Evaluation1. Conceptual Attention VariantsSingle-Head Attention: Forces all textual patterns to align onto a single contextual matrix. While useful for locating simple syntactic relationships, it performs poorly with homographs or global sentence meaning because its vector updates can only capture one context feature at a time.Multi-Head Attention: Splitting embeddings into parallel, distinct subspaces lets the model track multiple relationships at once. For example, Head 1 can resolve pronoun targets (coreference), Head 2 tracks verb objects, and Head 3 monitors sentence punctuation to find logical boundaries.Cross-Attention: Used in standard Encoder-Decoder platforms. Instead of generating $Q$, $K$, and $V$ matrices from the same sequence, the Queries ($Q$) come from the target translation/decoder sequence, while the Keys ($K$) and Values ($V$) come directly from the source encoder context. This aligns words across completely different sequences or modalities.2. Structural ReflectionBuilding a custom attention stack highlights the power of parallel matrix operations, but reveals a major limitation: Transformers lack an inductive bias for language structure. Without millions of training documents or pre-calculated positional encodings, an attention model treats text like an unarranged bag of words.Pre-trained baselines like DistilBERT bypass this bottleneck through transfer learning. Fine-tuning an existing model lets you apply pre-computed linguistic patterns directly to your specific dataset. This approach yields higher classification accuracy right out of the box compared to starting with raw, randomly initialized weights.
