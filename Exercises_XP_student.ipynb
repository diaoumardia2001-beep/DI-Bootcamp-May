What is the main difference between traditional machine learning and deep learning?
In traditional Machine Learning (ML), feature extraction (feature engineering) is manual: humans must select and build the most relevant variables to feed into the algorithm. In Deep Learning (DL), the model learns to automatically extract, filter, and structure these features directly from raw data through its successive layers.

How do Artificial Neural Networks (ANNs) mimic the human brain?
ANNs are biologically inspired by the brain's network of neurons. Artificial neurons (nodes) receive input signals, multiply them by specific weights (simulating the synaptic strength of dendrites), sum them up, and pass the signal through an activation function. If the signal exceeds a certain threshold, it is transmitted to the next layer (simulating an axon firing an electrical impulse).

Why does deep learning perform better on large datasets than traditional machine learning?
Traditional ML algorithms reach a performance plateau (saturation) where adding more data no longer improves accuracy. Deep Learning, thanks to its massive number of parameters and deep layered architecture, continuously discovers finer patterns and scales its accuracy almost linearly as the volume of data increases.

What are the challenges of deep learning, and how can they be addressed?

Overfitting (memorizing data): Solutions: Use Dropout layers, L1/L2 regularization, or data augmentation.

High Computational & Data Demands: Solutions: Use hardware accelerators (GPUs/TPUs) and leverage Transfer Learning (reusing pre-trained networks on large datasets).

The "Black Box" Problem (lack of explainability): Solutions: Use Explainable AI (XAI) frameworks like SHAP, LIME, or feature/activation maps (like Grad-CAM).

What is feature engineering, and why is it not necessary in deep learning?
Feature engineering is the process of manually transforming raw data into meaningful indicators to help an algorithm learn (e.g., extracting edges or textures to detect a cat). It is unnecessary in DL because the initial layers of a deep network naturally learn basic shapes (edges, corners), and deeper layers combine them into highly complex concepts (eyes, paws) on their own.

What role do hidden layers play in a deep learning model?
Hidden layers act as abstract feature extractors. As data propagates deeper through the hidden layers, the mathematical representations become increasingly complex and tailored to the specific target task.

In an Artificial Neural Network (ANN), what is the function of an activation function?
An activation function introduces non-linearity into the network. Without it, no matter how many hidden layers you stack, the network would reduce to a single linear combination (equivalent to a basic linear regression), making it incapable of learning complex patterns like images, audio, or text.


## 🌟 Exercise 2: Building a Simple Perceptron Decision System

**As stated in the instructions**  
You will create a perceptron-based decision system to decide whether you should go outside based on two inputs:
- Temperature in °F
- Rainy where Yes is 1 and No is 0

Given parameters: Temperature weight 0.6, Rain weight 0.4, Bias 2. Compute the weighted sum
\[ \text{Weighted Sum} = 0.6\cdot \text{Temperature} + 0.4\cdot \text{Rain} + \text{Bias} \]
Apply a step activation:
- If Weighted Sum > 20, output 1
- If Weighted Sum \le 20, output 0

Evaluate:  
Case 1: Temperature 70°F, Rain 0  
Case 2: Temperature 50°F, Rain 1

**Guidance**  
Implement the weighted sum function yourself. Use a reusable step activation. Then compute the two cases and print the decision for each.

**Learning point**  
A perceptron implements a linear decision boundary. The bias shifts the threshold. Use step activation for simple binary rules. Prefer differentiable activations when training with gradient descent.

![image.png](https://github.com/user-attachments/assets/3c4f3261-164d-4039-8e1c-aff2db7162dd)


```python
# PREFILLED: just execute
from typing import Tuple

W_TEMP = 0.6
W_RAIN = 0.4
BIAS   = 2.0

def step_activation(s: float, threshold: float = 20.0) -> int:
    """Return 1 if s > threshold else 0."""
    return 1 if s > threshold else 0
```

```python
# To-Do: implement weighted_sum then use it to compute the two cases
def weighted_sum(temperature_f: float, rain01: int) -> float:
    """Compute s = 0.6*Temperature + 0.4*Rain + Bias."""
    # To-Do: replace 'pass' with the formula using W_TEMP, W_RAIN, BIAS
    s = (W_TEMP * temperature_f) + (W_RAIN * rain01) + BIAS
    return s

# To-Do: compute decisions for the two cases and print them
case1 = (70, 0)  # Temperature=70, Rain=0
case2 = (50, 1)  # Temperature=50, Rain=1

s1 = weighted_sum(*case1)
s2 = weighted_sum(*case2)
y1 = step_activation(s1)
y2 = step_activation(s2)

print({"case1": {"sum": s1, "decision": y1}, "case2": {"sum": s2, "decision": y2}})
```

## 🌟 Exercise 5 (optional): Implementing Forward and Backward Propagation in Python

**As stated in the enoncé**  
You will code a simple neural network that performs forward propagation and backpropagation for a regression problem predicting exam scores from study hours.

**Guidance**  
Run the prefilled code. Then modify the learning rate or initial weights and observe changes in the loss and updates.
Provide a short written explanation.

**Learning point**  
Gradient descent updates parameters in the direction of steepest loss reduction.
The learning rate controls step size. Too small slows learning. Too large risks divergence.


```python
# PREFILLED: just execute
import numpy as np

# Initialize input data (features)
x = np.array([4, 80])  # 4 hours studied, previous test score: 80

# Initialize weights and bias
w = np.array([0.6, 0.3])  # Initial weights
b = 10  # Initial bias

# Forward Propagation
def forward_propagation(x, w, b):
    z = np.dot(x, w) + b  # Weighted sum
    return z  # Linear activation (No ReLU here, it's a regression task)

# Compute prediction
y_pred = forward_propagation(x, w, b)
y_true = 85  # Actual exam score

# Compute Loss (Mean Squared Error)
loss = 0.5 * (y_true - y_pred) ** 2

# Compute Gradients
grad_w = -(y_true - y_pred) * x  # Partial derivatives with respect to weights
grad_b = -(y_true - y_pred)  # Partial derivative with respect to bias

# Update Weights and Bias
learning_rate = 0.01
w_new = w - learning_rate * grad_w
b_new = b - learning_rate * grad_b

# Print Results
print("Initial Prediction:", y_pred)
print("Loss:", loss)
print("Updated Weights:", w_new)
print("Updated Bias:", b_new)
```

**To-Do:** Explain in full sentences why updating weights using gradient descent reduces the error. Then modify the learning rate or initial weights and describe how this affects learning.
1. Why Gradient Descent Reduces the Error
Updating weights using gradient descent reduces prediction error because it relies on the partial derivatives of the loss function, which mathematically define the slope of the error curve. The gradient vector always points in the direction of the steepest increase in error. By calculating this gradient and subtracting it from the current weights, we force the parameters to move in the exact opposite direction—downward toward the lowest point of the error surface, known as the local or global minimum. As the weights slide down this curve closer to the minimum, the mathematical discrepancy between the model's predictions and the true target values systematically decreases.

2. Modifying the Parameters and Their Effects on Learning
Scenario A: Increasing the Learning Rate (e.g., learning_rate = 0.05)
If you significantly increase the learning rate, the step sizes taken down the error curve become too large. Instead of settling smoothly into the minimum, the updates will aggressively overshoot the optimal weights. This causes the model's predictions to wildly oscillate back and forth across the target value, leading to a phenomenon where the loss violently destabilizes or explodes entirely rather than converging.

Scenario B: Decreasing the Learning Rate (e.g., learning_rate = 0.0001)
If you lower the learning rate to an extremely small value, the weight updates become incredibly minute. While the model will still technically move in the correct direction toward minimizing the error, its progress will be exceptionally slow. The network will require an immense number of training iterations (epochs) to achieve accuracy, rendering the learning process computationally inefficient and prone to getting stuck in flat regions of the error space.

Scenario C: Altering Initial Weights (e.g., w = np.array([10.0, -5.0]))
Changing the initial weights alters the starting coordinate of the model on the error surface. If you initialize the weights much further away from the optimal values, the initial prediction error and loss will spike significantly higher. Consequently, the model will require more optimization steps to clear the distance to the minimum. Conversely, starting with weights that happen to be closer to the ideal ratios allows the network to converge much faster.
