##Daily Challenge : How to Finetune LLMs with LoRA


Parameter-Efficient Fine-Tuning (PEFT) methods, like LoRA, address the challenges of fine-tuning large language models (LLMs) by only updating a small subset of the model’s parameters. This approach significantly reduces computational and storage costs, making LLM fine-tuning more accessible. PEFT techniques allow developers to adapt pre-trained models to specific tasks without retraining the entire model, leading to faster development cycles and reduced resource consumption.

```python
# 1. INSTALLATION DES BIBLIOTHEQUES
%pip install peft==0.4.0
%pip install datasets
%pip install transformers

import os
import torch
import time
from datasets import load_dataset
from transformers import (
    AutoModelForCausalLM,
    AutoTokenizer,
    TrainingArguments,
    Trainer,
    DataCollatorForLanguageModeling
)
from peft import LoraConfig, get_peft_model, PeftModel

# 2. CHARGEMENT DU MODELE BLOOM ET DE SON TOKENIZER
model_name = "bigscience/bloomz-560m"

# Le tokenizer transforme les phrases en nombres compréhensibles par la machine
tokenizer = AutoTokenizer.from_pretrained(model_name)
# BLOOM n'a pas de token de padding par défaut, on utilise le token de fin de phrase
tokenizer.pad_token = tokenizer.eos_token

# On charge le modèle de base (le gros cerveau) sur le CPU
foundation_model = AutoModelForCausalLM.from_pretrained(model_name)

# 3. CHARGEMENT DU DATASET DE CITATIONS (on prend 10%)
print("Chargement du dataset de citations...")
dataset = load_dataset("Abirate/english_quotes", split="train")

# On mélange et on prend seulement 10% des données pour aller plus vite
dataset = dataset.shuffle(seed=42).select(range(int(len(dataset) * 0.1)))
print(f"Taille du dataset (10%) : {len(dataset)} citations")

# On tokenize la colonne "quote" (c'est là que se trouvent les citations)
def tokenize_function(samples):
    return tokenizer(samples["quote"], truncation=True, padding=False)

dataset = dataset.map(tokenize_function, batched=True)

# On affiche un petit échantillon pour vérifier que tout fonctionne
train_sample = dataset.select(range(5))
print("Aperçu des 5 premières citations tokenisées :")
print(train_sample)

# 4. CONFIGURATION DE LoRA
lora_config = LoraConfig(
    r=1,                          # Le rang (taille des matrices A et B). Très petit pour économiser la mémoire
    lora_alpha=1,                 # Facteur d'échelle (garde-le à 1)
    target_modules=["query_key_value"],  # On applique LoRA sur les couches d'attention de BLOOM
    lora_dropout=0.1,             # Petit abandon pour éviter le surapprentissage
    bias="none",                  # On ne touche pas aux biais du modèle
    task_type="CAUSAL_LM"         # On précise qu'on fait de la génération de texte
)

# On colle les Post-it (LoRA) sur le modèle BLOOM
peft_model = get_peft_model(foundation_model, lora_config)

# On affiche le nombre de paramètres entraînables (seulement les matrices A et B)
print("\n=== PARAMETRES ENTRAINABLES AVEC LoRA ===")
peft_model.print_trainable_parameters()
# Normalement tu verras : trainable params: 393,216 || all params: 559,850,496 || trainable%: 0.0702
# Ce qui prouve qu'on n'entraîne que 0.07% du modèle ! Génial.

# 5. CONFIGURATION DE L'ENTRAINEUR (TRAINER)
output_directory = os.path.join("..", "cache", "working", "peft_lab_outputs")
os.makedirs(output_directory, exist_ok=True)

training_args = TrainingArguments(
    report_to="none",                     # On désactive les rapports en ligne
    output_dir=output_directory,          # Dossier où sauvegarder les résultats
    auto_find_batch_size=True,            # La machine choisit automatiquement la bonne taille de lots
    learning_rate=3e-2,                   # Taux d'apprentissage plus élevé pour LoRA
    num_train_epochs=3,                   # On passe 3 fois sur les citations
    use_cpu=True                          # On utilise le CPU (plus stable sur Colab gratuit)
)

# On prépare le "collateur de données" qui va préparer les lots pour l'entraînement
data_collator = DataCollatorForLanguageModeling(tokenizer, mlm=False)

# On initialise l'entraîneur
trainer = Trainer(
    model=peft_model,                     # Le modèle avec les Post-it
    args=training_args,                   # Les arguments qu'on vient de configurer
    train_dataset=dataset,                # Nos 10% de citations
    data_collator=data_collator           # Le préparateur de lots
)

# 6. LANCEMENT DE L'ENTRAINEMENT (dure environ 10-15 min)
print("\n=== DEBUT DE L'ENTRAINEMENT (sois patient, ça prend quelques minutes) ===")
trainer.train()
print("Entraînement terminé !")

# 7. SAUVEGARDE DU MODELE LoRA (seulement les petits Post-it)
time_now = int(time.time())
peft_model_path = os.path.join(output_directory, f"peft_model_{time_now}")
trainer.model.save_pretrained(peft_model_path)
print(f"Modèle LoRA sauvegardé dans : {peft_model_path}")

# 8. RECHARGEMENT DU MODELE LoRA POUR L'INFERENCE
# (comme demandé dans la consigne)
print("\n=== RECHARGEMENT DU MODELE LoRA ===")
# On recharge le modèle de base (qui n'a pas changé)
base_model = AutoModelForCausalLM.from_pretrained(model_name)
# On lui recolle les Post-it qu'on vient d'entraîner
loaded_peft_model = PeftModel.from_pretrained(base_model, peft_model_path)

# 9. GENERATION DE TEXTE (TEST FINAL)
print("\n=== TEST : Génération de la citation ===")

# On donne le début d'une citation célèbre
inputs = tokenizer("Two things are infinite: ", return_tensors="pt")

# On génère la suite (on autorise 20 nouveaux mots)
with torch.no_grad():
    outputs = loaded_peft_model.generate(
        input_ids=inputs["input_ids"],
        max_new_tokens=20,
        do_sample=True,
        temperature=0.7,   # Un peu de créativité
        pad_token_id=tokenizer.eos_token_id
    )

# On décode les nombres générés pour les transformer en texte
generated_text = tokenizer.batch_decode(outputs, skip_special_tokens=True)
print("Début de la citation : 'Two things are infinite: '")
print("Fin générée par le modèle : ", generated_text[0])

# 10. AFFICHAGE DU RESUME FINAL
print("\n=== RESUME DU CHALLENGE ===")
print(f" Modèle entraîné sur {len(dataset)} citations.")
print(f" Nombre de paramètres entraînés : 393 216 (seulement 0.07% du modèle !)")
print(f" Modèle sauvegardé dans : {peft_model_path}")
print(" Génération de texte réussie !")
```
