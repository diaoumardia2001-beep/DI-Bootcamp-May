Solving the frequency problem using TF-IDF
When we create a BoW out from some text, all the words are treated equaly as importants. Like “Alice” or “say” in a fantasy book about Alice. We expected those words to be repeated all over the book, making them not so informative to us.

The solution for this problem would be to consider the frequency relative to the corpus. In this case, if there is a word in a document that doesn’t appears much in the other documents, it is likely meaningful and should be considered more important. And the same way in the oposite: A word that is repeated a lot in all the documents will be considered less important.

That’s the perfect situation to use TF-IDF (Term Frequency-Inverse Document Frequency)

```python
import re
import requests
import pandas as pd
import spacy
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import nltk
nltk.download('punkt_tab') # Add this line to download the missing resource

# 1. Chargement des données depuis Gutenberg
urls = [
    "https://www.gutenberg.org/cache/epub/11/pg11.txt",      # Alice's Adventures in Wonderland
    "https://www.gutenberg.org/cache/epub/12/pg12.txt",      # Through the Looking-Glass
    "https://www.gutenberg.org/cache/epub/29042/pg29042.txt" # A Tangled Tale
]

corpus = []
titles = ["Alice in Wonderland", "Through the Looking-Glass", "A Tangled Tale"]
nlp = spacy.load('en_core_web_sm')
stop_words = set(stopwords.words('english'))

for url, title in zip(urls, titles):
    response = requests.get(url)
    response.encoding = 'utf-8'
    text = response.text

    # Slicing pour retirer l'en-tête et le pied de page Gutenberg
    start_match = re.search(r"\*\*\* START OF THIS PROJECT[^*]*\*\*\*", text)
    end_match = re.search(r"\*\*\* END OF THIS PROJECT[^*]*\*\*\*", text)
    if start_match and end_match:
        text = text[start_match.end():end_match.start()]

    # Nettoyage regex (uniquement lettres et espaces)
    text_cleaned = re.sub(r'[^a-zA-Z\s]', '', text)
    text_cleaned = re.sub(r'\s+', ' ', text_cleaned).strip().lower()

    # Tokenisation et retrait des Stopwords
    tokens = word_tokenize(text_cleaned)
    tokens_no_stop = [t for t in tokens if t not in stop_words]

    # Lemmatisation complète (via spaCy)
    # Colab gère de gros volumes, mais on limite par sécurité pour accélérer l'exécution
    clean_str = " ".join(tokens_no_stop[:15000])
    doc = nlp(clean_str)
    final_text = " ".join([token.lemma_ for token in doc])

    corpus.append({"title": title, "final_cleaned_text": final_text})

print(" Textes chargés et nettoyés avec succès !")
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import matplotlib.pyplot as plt # Import matplotlib.pyplot for plotting

documents = [book['final_cleaned_text'] for book in corpus]
book_titles = [book['title'] for book in corpus]

# --- 1. APPROCHE BAG OF WORDS ---
vectorizer_bow = CountVectorizer()
bow_matrix = vectorizer_bow.fit_transform(documents)
df_bow = pd.DataFrame(bow_matrix.toarray(), columns=vectorizer_bow.get_feature_names_out(), index=book_titles)

# --- 2. APPROCHE TF-IDF ---
vectorizer_tfidf = TfidfVectorizer(min_df=1, max_df=2)
tfidf_matrix = vectorizer_tfidf.fit_transform(documents)
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=vectorizer_tfidf.get_feature_names_out(), index=book_titles)

# --- 3. AFFICHAGE DES GRAPHIQUES COMPARATIFS ---
fig, axes = plt.subplots(2, 3, figsize=(18, 12))

colors_bow = ['#6366F1', '#34D399', '#F59E0B', '#EF4444', '#10B981']
colors_tfidf = ['#8B5CF6', '#EC4899', '#3B82F6', '#F59E0B', '#10B981']

for idx, title in enumerate(book_titles):
    # Top 5 Mots - Bag of Words
    top5_bow = df_bow.loc[title].sort_values(ascending=False).head(5)
    axes[0, idx].pie(top5_bow.values, labels=top5_bow.index, autopct='%1.1f%%', startangle=140, colors=colors_bow)
    axes[0, idx].set_title(f" BoW : {title}", fontsize=12, weight='bold')

    # Top 5 Mots - TF-IDF
    top5_tfidf = df_tfidf.loc[title].sort_values(ascending=False).head(5)
    axes[1, idx].pie(top5_tfidf.values, labels=top5_tfidf.index, autopct='%1.1f%%', startangle=140, colors=colors_tfidf)
    axes[1, idx].set_title(f" TF-IDF : {title}", fontsize=12, weight='bold')

plt.suptitle("Comparaison de l'importance des mots : BoW vs TF-IDF", fontsize=16, weight='bold', y=0.98)
plt.tight_layout()
plt.show()
```
