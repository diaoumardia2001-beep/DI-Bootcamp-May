```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models

# Load the built-in MNIST handwritten digits dataset
(X_train_raw, y_train_raw), (X_test_raw, y_test_raw) = tf.keras.datasets.mnist.load_data()

print("=== RAW DATASET STRUCTURE ===")
print(f"Training Images Shape : {X_train_raw.shape}  -> (60,000 samples, 28x28 pixels)")
print(f"Training Labels Shape : {y_train_raw.shape}  -> (60,000 ground truth integers)")
print(f"Testing Images Shape  : {X_test_raw.shape}  -> (10,000 samples, 28x28 pixels)")
print(f"Testing Labels Shape  : {y_test_raw.shape}  -> (10,000 ground truth integers)")
```

```python
print("\n=== PREPROCESSING FOR FULLY CONNECTED NETWORK ===")

# 1. Flatten the spatial matrices (28x28 -> 784)
X_train_dense = X_train_raw.reshape(-1, 784).astype("float32")
X_test_dense = X_test_raw.reshape(-1, 784).astype("float32")

# 2. Max-Min Feature Scaling Normalization [0, 255] -> [0.0, 1.0]
X_train_dense /= 255.0
X_test_dense /= 255.0

# 3. Categorical One-Hot Target Encoding
num_classes = 10
y_train_categorical = tf.keras.utils.to_categorical(y_train_raw, num_classes)
y_test_categorical = tf.keras.utils.to_categorical(y_test_raw, num_classes)

print(f"Dense Train Input Shape: {X_train_dense.shape}")
print(f"Dense Train Label Shape: {y_train_categorical.shape}")

# 4. Build the Fully Connected Architecture
mlp_model = models.Sequential([
    layers.Input(shape=(784,)),
    layers.Dense(512, activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(num_classes, activation='softmax')
])

mlp_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

mlp_model.summary()

# 5. Train the Model
print("\n Training Fully Connected (Dense) Network...")
mlp_history = mlp_model.fit(
    X_train_dense, y_train_categorical,
    validation_data=(X_test_dense, y_test_categorical),
    epochs=5,
    batch_size=128,
    verbose=1
)

# Evaluate final test accuracy
mlp_loss, mlp_acc = mlp_model.evaluate(X_test_dense, y_test_categorical, verbose=0)
print(f"\n Dense Network Final Test Accuracy: {mlp_acc * 100:.2f}%")
```

```python
print("\n=== PREPROCESSING FOR CONVOLUTIONAL NEURAL NETWORK ===")

# 1. Reshape inputs to include the single grayscale channel dimension (Height, Width, Channels)
X_train_cnn = X_train_raw.reshape(-1, 28, 28, 1).astype("float32")
X_test_cnn = X_test_raw.reshape(-1, 28, 28, 1).astype("float32")

# 2. Normalize Pixel Values
X_train_cnn /= 255.0
X_test_cnn /= 255.0

print(f"CNN Train Input Shape: {X_train_cnn.shape}")

# 3. Build the Convolutional Architecture
cnn_model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),

    # Feature extraction blocks (Convolutions + Pooling)
    layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    layers.MaxPooling2D((2, 2)),

    # Transition to structural classification head
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(num_classes, activation='softmax')
])

cnn_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

cnn_model.summary()

# 4. Train the Convolutional Model
print("\n Training Convolutional Neural Network (CNN)...")
cnn_history = cnn_model.fit(
    X_train_cnn, y_train_categorical,
    validation_data=(X_test_cnn, y_test_categorical),
    epochs=5,
    batch_size=128,
    verbose=1
)

# Evaluate final test accuracy
cnn_loss, cnn_acc = cnn_model.evaluate(X_test_cnn, y_test_categorical, verbose=0)
print(f"\n CNN Network Final Test Accuracy: {cnn_acc * 100:.2f}%")
```

When you execute these runs, you will observe a distinct gap in performance metrics. Here is why the two models diverge:Comparison BreakdownParameter Efficiency vs. Raw Count: Look at your model summaries. The Fully Connected Network actually handles an immense number of connections because every single pixel input has its own unique weight leading to the next hidden layer. The CNN uses far fewer parameters in its feature layers due to weight sharing—the same small $3 \times 3$ sliding window filter is reused across the entire image space.Spatial Blindness: The Dense network treats pixel #1 and pixel #2 with the exact same relational mechanics as pixel #1 and pixel #784. If you shift a handwritten digit just 3 pixels to the left, a Dense network completely panics because its input slots change positions entirely.Spatial Invariance (The CNN Edge): The Convolutional layers look for localized geometric primitives (edges, curves, loops) regardless of where they reside on the canvas. This is why your CNN comfortably crosses into the $98\text{--}99\%$ validation accuracy territory within a few short epochs, while the Dense network plateaus early.
