The Pareto Principle, also known as the 80/20 rule, is a concept derived from the work of Italian economist Vilfredo Pareto. It states that roughly 80% of the effects come from 20% of the causes. For instance, identifying the top 20% of products that generate 80% of sales or the top 20% of customers that contribute to 80% of profit can help in prioritizing efforts and resources. This focus can lead to improved efficiency and effectiveness in business strategies. Can we apply Pareto principle to customers and Profit ? (Determine if 20% of the customers contribute to 80% of the profit.)
What are the Top 20 cities by Sales ? What about the Top 20 cities by Profit ? Are there any difference among cities in profitability ? (Identify the top 20 cities based on total sales and total profit and analyze differences in profitability among these cities.)
What are the Top 20 customers by Sales?
Plot the Cumulative curve in Sales by Customers. Can we apply Pareto principle to customers and Sales ?
Based on the analysis, make decisions on which states and cities to prioritize for marketing strategies.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import requests
import io
import sys
!{sys.executable} -m pip install xlrd
import warnings
warnings.filterwarnings('ignore')

# ============================================================
# LOAD AND PREPROCESS DATASET
# ============================================================

url = "https://github.com/devtlv/Datasets-DA-Bootcamp-2-/raw/refs/heads/main/Week%205%20-%20Data%20Processing/W5D5%20-%20Mini-project%20-%20bis/US%20Superstore%20data.xls"
response = requests.get(url)
df = pd.read_excel(r"C:\Users\Utilisateur\OneDrive\Desktop\DI-Bootcamp-May\Week2\Day5\Exercice\US Superstore data.xls")
print("Dataset Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst rows:")
print(df.head())
print("\nMissing values:")
print(df.isnull().sum())

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date']  = pd.to_datetime(df['Ship Date'])
df['Order Year'] = df['Order Date'].dt.year
df['Order Month']= df['Order Date'].dt.month
df = df.drop_duplicates()

print("\nData types after preprocessing:")
print(df.dtypes)

# ============================================================
# QUESTION 1 : Which states have the most sales?
# ============================================================

state_sales = df.groupby('State')['Sales'].sum().sort_values(ascending=False).head(20)

plt.figure(figsize=(14, 8))
bars = plt.barh(state_sales.index[::-1], state_sales.values[::-1],
                color='steelblue', edgecolor='black', alpha=0.8)
plt.title('Top 20 States by Total Sales', fontsize=16, fontweight='bold')
plt.xlabel('Total Sales ($)', fontsize=12)
plt.ylabel('State', fontsize=12)
for bar, value in zip(bars, state_sales.values[::-1]):
    plt.text(value + 1000, bar.get_y() + bar.get_height()/2,
            f'${value:,.0f}', va='center', fontsize=9)
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

print("Top 10 States by Sales:")
print(state_sales.head(10))

# ============================================================
# QUESTION 2 : NY vs California - Sales and Profit
# ============================================================

ny_ca = df[df['State'].isin(['New York', 'California'])]
ny_ca_summary = ny_ca.groupby('State')[['Sales', 'Profit']].sum()

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('New York vs California - Sales & Profit Comparison',
            fontsize=16, fontweight='bold')

axes[0].bar(ny_ca_summary.index, ny_ca_summary['Sales'],
            color=['steelblue', 'orange'], edgecolor='black', alpha=0.8)
axes[0].set_title('Total Sales', fontsize=14)
axes[0].set_ylabel('Sales ($)', fontsize=12)
for i, (state, row) in enumerate(ny_ca_summary.iterrows()):
    axes[0].text(i, row['Sales'] + 1000, f'${row["Sales"]:,.0f}',
                ha='center', fontweight='bold')
axes[0].grid(axis='y', alpha=0.3)

axes[1].bar(ny_ca_summary.index, ny_ca_summary['Profit'],
            color=['steelblue', 'orange'], edgecolor='black', alpha=0.8)
axes[1].set_title('Total Profit', fontsize=14)
axes[1].set_ylabel('Profit ($)', fontsize=12)
for i, (state, row) in enumerate(ny_ca_summary.iterrows()):
    axes[1].text(i, row['Profit'] + 500, f'${row["Profit"]:,.0f}',
                ha='center', fontweight='bold')
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\nNew York vs California Summary:")
print(ny_ca_summary)

# ============================================================
# QUESTION 3 : Outstanding customer in New York
# ============================================================

ny_customers = df[df['State'] == 'New York'].groupby('Customer Name')[['Sales', 'Profit']].sum()
ny_customers = ny_customers.sort_values('Sales', ascending=False).head(10)

plt.figure(figsize=(12, 6))
bars = plt.barh(ny_customers.index[::-1], ny_customers['Sales'][::-1],
                color='steelblue', edgecolor='black', alpha=0.8)
plt.title('Top 10 Customers in New York by Sales', fontsize=16, fontweight='bold')
plt.xlabel('Total Sales ($)', fontsize=12)
plt.ylabel('Customer Name', fontsize=12)
for bar, value in zip(bars, ny_customers['Sales'][::-1]):
    plt.text(value + 100, bar.get_y() + bar.get_height()/2,
            f'${value:,.0f}', va='center', fontsize=9)
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

print("\nTop Customer in New York:")
print(f"  Name  : {ny_customers.index[0]}")
print(f"  Sales : ${ny_customers['Sales'].iloc[0]:,.0f}")
print(f"  Profit: ${ny_customers['Profit'].iloc[0]:,.0f}")

# ============================================================
# QUESTION 4 : Differences among states in profitability
# ============================================================

state_profit = df.groupby('State')['Profit'].sum().sort_values(ascending=False)
colors = ['green' if x > 0 else 'red' for x in state_profit.values]

plt.figure(figsize=(16, 10))
plt.bar(range(len(state_profit)), state_profit.values, color=colors, edgecolor='black', alpha=0.8)
plt.xticks(range(len(state_profit)), state_profit.index, rotation=90, fontsize=8)
plt.axhline(y=0, color='black', linewidth=1.5)
plt.title('Profitability by State (Green=Profit, Red=Loss)', fontsize=16, fontweight='bold')
plt.xlabel('State', fontsize=12)
plt.ylabel('Total Profit ($)', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

losing_states = state_profit[state_profit < 0]
print(f"\nStates with losses ({len(losing_states)}):")
print(losing_states)

# ============================================================
# QUESTION 5 : Pareto Principle - Customers and Profit
# ============================================================

customer_profit = df.groupby('Customer Name')['Profit'].sum().sort_values(ascending=False)
customer_profit = customer_profit[customer_profit > 0]

cumulative_profit  = customer_profit.cumsum()
total_profit       = customer_profit.sum()
cumulative_pct     = (cumulative_profit / total_profit) * 100
customer_pct       = np.arange(1, len(customer_profit) + 1) / len(customer_profit) * 100

pareto_idx = (cumulative_pct >= 80).idxmax()
pareto_pct = customer_pct[customer_profit.index.get_loc(pareto_idx)]

plt.figure(figsize=(12, 6))
plt.plot(customer_pct, cumulative_pct, color='steelblue', linewidth=2)
plt.axhline(y=80, color='red', linestyle='--', linewidth=1.5, label='80% Profit')
plt.axvline(x=pareto_pct, color='orange', linestyle='--', linewidth=1.5,
            label=f'{pareto_pct:.1f}% Customers')
plt.fill_between(customer_pct, cumulative_pct, alpha=0.1, color='steelblue')
plt.title('Pareto Principle - Customers vs Profit', fontsize=16, fontweight='bold')
plt.xlabel('% of Customers', fontsize=12)
plt.ylabel('Cumulative % of Profit', fontsize=12)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"\nPareto Analysis - Customers & Profit:")
print(f"  {pareto_pct:.1f}% of customers generate 80% of profit")
if pareto_pct <= 20:
    print("  Pareto Principle CONFIRMED (<=20% customers = 80% profit)")
else:
    print(f"  Pareto Principle NOT exactly confirmed ({pareto_pct:.1f}% needed)")

# ============================================================
# QUESTION 6 : Top 20 Cities by Sales and Profit
# ============================================================

top20_cities_sales  = df.groupby('City')['Sales'].sum().sort_values(ascending=False).head(20)
top20_cities_profit = df.groupby('City')['Profit'].sum().sort_values(ascending=False).head(20)

fig, axes = plt.subplots(1, 2, figsize=(18, 8))
fig.suptitle('Top 20 Cities - Sales vs Profit', fontsize=16, fontweight='bold')

axes[0].barh(top20_cities_sales.index[::-1], top20_cities_sales.values[::-1],
            color='steelblue', edgecolor='black', alpha=0.8)
axes[0].set_title('Top 20 Cities by Sales', fontsize=14)
axes[0].set_xlabel('Total Sales ($)', fontsize=12)
axes[0].grid(axis='x', alpha=0.3)

colors_profit = ['green' if x > 0 else 'red' for x in top20_cities_profit.values[::-1]]
axes[1].barh(top20_cities_profit.index[::-1], top20_cities_profit.values[::-1],
            color=colors_profit, edgecolor='black', alpha=0.8)
axes[1].set_title('Top 20 Cities by Profit', fontsize=14)
axes[1].set_xlabel('Total Profit ($)', fontsize=12)
axes[1].axvline(x=0, color='black', linewidth=1)
axes[1].grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.show()

city_both = df.groupby('City')[['Sales', 'Profit']].sum()
city_both['Profit Margin'] = (city_both['Profit'] / city_both['Sales']) * 100
print("\nTop 20 Cities - Sales, Profit and Margin:")
print(city_both.sort_values('Sales', ascending=False).head(20))

# ============================================================
# QUESTION 7 : Top 20 Customers by Sales
# ============================================================

top20_customers = df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(20)

plt.figure(figsize=(12, 8))
plt.barh(top20_customers.index[::-1], top20_customers.values[::-1],
        color='steelblue', edgecolor='black', alpha=0.8)
plt.title('Top 20 Customers by Sales', fontsize=16, fontweight='bold')
plt.xlabel('Total Sales ($)', fontsize=12)
plt.ylabel('Customer Name', fontsize=12)
for i, value in enumerate(top20_customers.values[::-1]):
    plt.text(value + 100, i, f'${value:,.0f}', va='center', fontsize=9)
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

print("\nTop 5 Customers by Sales:")
print(top20_customers.head())

# ============================================================
# QUESTION 8 : Cumulative Sales Curve + Pareto (Customers)
# ============================================================

customer_sales    = df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False)
cumulative_sales  = customer_sales.cumsum()
total_sales       = customer_sales.sum()
cumulative_pct_s  = (cumulative_sales / total_sales) * 100
customer_pct_s    = np.arange(1, len(customer_sales) + 1) / len(customer_sales) * 100

pareto_idx_s  = (cumulative_pct_s >= 80).idxmax()
pareto_pct_s  = customer_pct_s[customer_sales.index.get_loc(pareto_idx_s)]

plt.figure(figsize=(12, 6))
plt.plot(customer_pct_s, cumulative_pct_s, color='steelblue', linewidth=2,
        label='Cumulative Sales')
plt.axhline(y=80, color='red', linestyle='--', linewidth=1.5, label='80% Sales')
plt.axvline(x=pareto_pct_s, color='orange', linestyle='--', linewidth=1.5,
            label=f'{pareto_pct_s:.1f}% Customers')
plt.fill_between(customer_pct_s, cumulative_pct_s, alpha=0.1, color='steelblue')
plt.title('Cumulative Sales Curve by Customers (Pareto Analysis)',
        fontsize=16, fontweight='bold')
plt.xlabel('% of Customers', fontsize=12)
plt.ylabel('Cumulative % of Sales', fontsize=12)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"\nPareto Analysis - Customers & Sales:")
print(f"  {pareto_pct_s:.1f}% of customers generate 80% of sales")
if pareto_pct_s <= 20:
    print("  Pareto Principle CONFIRMED")
else:
    print(f"  Pareto Principle NOT exactly confirmed ({pareto_pct_s:.1f}% needed)")

# ============================================================
# QUESTION 9 : Marketing Strategy Recommendations
# ============================================================

print("\n" + "=" * 60)
print("      MARKETING STRATEGY RECOMMENDATIONS")
print("=" * 60)

top5_states  = df.groupby('State')['Sales'].sum().sort_values(ascending=False).head(5)
top5_cities  = df.groupby('City')['Sales'].sum().sort_values(ascending=False).head(5)
top5_profit_states = df.groupby('State')['Profit'].sum().sort_values(ascending=False).head(5)

print(f"\nPRIORITY STATES FOR MARKETING:")
for state, sales in top5_states.items():
    profit = df[df['State'] == state]['Profit'].sum()
    print(f"  {state:<20} Sales: ${sales:>10,.0f}  Profit: ${profit:>10,.0f}")

print(f"\nPRIORITY CITIES FOR MARKETING:")
for city, sales in top5_cities.items():
    profit = df[df['City'] == city]['Profit'].sum()
    print(f"  {city:<20} Sales: ${sales:>10,.0f}  Profit: ${profit:>10,.0f}")

print(f"\nSTATES TO AVOID OR REVIEW (Losses):")
losing = df.groupby('State')['Profit'].sum().sort_values().head(5)
for state, profit in losing.items():
    print(f"  {state:<20} Profit: ${profit:>10,.0f}")

print(f"\nKEY RECOMMENDATIONS:")
print(f"  1. Focus marketing budget on top 5 states by sales")
print(f"  2. Prioritize top 5 cities for targeted campaigns")
print(f"  3. Review discount strategy in loss-making states")
print(f"  4. Retain top 20% customers who drive 80% of profit")
print(f"  5. Investigate and fix unprofitable product categories")
print("=" * 60)
```
