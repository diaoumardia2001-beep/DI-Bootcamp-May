## Exercise 1: Understanding Data Visualization
Task: Explain why data visualization is important in data analysis.
Task: Describe the purpose of a line graph in data visualization.


```python
# Exercise 1: Understanding Data Visualization

# ============================================================
# TASK 1 : Why is data visualization important in data analysis?
# ============================================================

print("=== WHY DATA VISUALIZATION IS IMPORTANT ===\n")

print("""
Data visualization is important in data analysis for several reasons:

1. SIMPLIFIES COMPLEX DATA
   Raw data is often large and difficult to interpret.
   Visualization transforms numbers into charts and graphs
   that are easier to understand at a glance.

2. REVEALS PATTERNS AND TRENDS
   Visualizations help identify trends, patterns, and
   correlations that would be hard to spot in raw tables.
   Example: A line chart instantly shows if sales are
   increasing or decreasing over time.

3. FASTER DECISION MAKING
   Business leaders can make faster and better decisions
   when data is presented visually rather than in spreadsheets.

4. COMMUNICATES INSIGHTS CLEARLY
   Visualizations make it easier to communicate findings
   to non-technical stakeholders who may not understand
   complex statistical outputs.

5. IDENTIFIES OUTLIERS AND ANOMALIES
   Charts make it easy to spot unusual data points
   (outliers) that might indicate errors or opportunities.

6. TELLS A STORY
   Good visualizations guide the audience through the data
   narrative, making the analysis more memorable and impactful.
""")

# ============================================================
# TASK 2 : What is the purpose of a line graph?
# ============================================================

print("=== PURPOSE OF A LINE GRAPH ===\n")

print("""
A line graph is used to:

1. SHOW TRENDS OVER TIME
   It displays how a variable changes continuously over
   a period (days, months, years).
   Example: Monthly sales from 2020 to 2023.

2. COMPARE MULTIPLE SERIES
   Multiple lines on the same chart allow comparison
   of different categories over the same time period.
   Example: Sales vs Profit over 12 months.

3. SHOW CONTINUITY
   Unlike bar charts, line graphs emphasize the
   continuous nature of data and the connection
   between data points.

4. IDENTIFY SEASONALITY
   Line graphs make it easy to spot recurring patterns
   or seasonal fluctuations in the data.
   Example: Sales peaks every December (holiday season).

5. DETECT ANOMALIES
   Sudden spikes or drops in the line clearly indicate
   unusual events or outliers in the data.
""")

# ============================================================
# EXAMPLE : Line graph vs Bar chart
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
         'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales  = [1200, 1500, 1300, 1700, 1600, 1900,
         2100, 1800, 2000, 2300, 2500, 2800]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(months, sales, marker='o', color='steelblue',
      linewidth=2, markersize=6)
axes[0].set_title('Line Graph - Monthly Sales Trend',
                  fontsize=14, fontweight='bold')
axes[0].set_xlabel('Month')
axes[0].set_ylabel('Sales ($)')
axes[0].grid(True, alpha=0.3)

axes[1].bar(months, sales, color='steelblue', edgecolor='black', alpha=0.7)
axes[1].set_title('Bar Chart - Monthly Sales',
                  fontsize=14, fontweight='bold')
axes[1].set_xlabel('Month')
axes[1].set_ylabel('Sales ($)')
axes[1].grid(axis='y', alpha=0.3)

plt.suptitle('Line Graph vs Bar Chart - Same Data, Different Purpose',
fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()

print("Conclusion:")
print("  Line graph → best for showing TRENDS over time")
print("  Bar chart  → best for COMPARING values at specific points")
```

## Exercise 2: Creating a Line Plot for Temperature Variation
Objective: Create a simple line plot using Matplotlib that represents temperature variations over a week.
Tasks:
Use a list of temperature values for each day of the week (e.g., [72, 74, 76, 80, 82, 78, 75]).
Label the x-axis as “Day” and the y-axis as “Temperature (°F)”.
Add a title to the plot.
Display the plot.


```python
import matplotlib.pyplot as plt

days         = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
temperatures = [72, 74, 76, 80, 82, 78, 75]

plt.figure(figsize=(10, 6))

plt.plot(days, temperatures, marker='o', color='steelblue', linewidth=2, markersize=8)

plt.title('Temperature Variation Over a Week', fontsize=16, fontweight='bold')
plt.xlabel('Day', fontsize=12)
plt.ylabel('Temperature (°F)', fontsize=12)

for i, temp in enumerate(temperatures):
    plt.text(i, temp + 0.3, f'{temp}°F', ha='center', fontsize=10)

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Exercise 3: Visualizing Monthly Sales with a Bar Chart
Objective: Generate a bar chart using Matplotlib to visualize monthly sales data for a retail store.
Tasks:
Create a list of sales values for each month (e.g., [5000, 5500, 6200, 7000, 7500]).
Label the x-axis as “Month” and the y-axis as “Sales Amount ($)”.
Add a title to the bar chart.
Display the plot.

```python
import matplotlib.pyplot as plt

months = ['January', 'February', 'March', 'April', 'May']
sales  = [5000, 5500, 6200, 7000, 7500]

plt.figure(figsize=(10, 6))

bars = plt.bar(months, sales, color='steelblue', edgecolor='black', alpha=0.8)

plt.title('Monthly Sales - Retail Store', fontsize=16, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Sales Amount ($)', fontsize=12)

for bar, value in zip(bars, sales):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 50,
        f'${value:,}',
        ha='center',
        fontsize=10,
        fontweight='bold'
    )

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## Exercise 6: Exploring the Relationship Between Age and Panic Attacks
Objective: Create a scatter plot to explore the relationship between students’ age and the occurrence of panic attacks.
Dataset Overview: The dataset includes columns: ‘Age’ and ‘Do you have Panic Attacks?’.
Tasks:
Import necessary libraries.
Convert panic attack responses to numeric values (e.g., Yes=1, No=0).
Use Seaborn’s scatterplot to create a scatter plot with ‘Age’ on the x-axis and numeric panic attack responses on the y-axis.
Customize the plot to improve readability by adding labels, a title, and adjusting point styles.
Display the plot.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import requests
import zipfile
import io

url = "https://github.com/devtlv/Datasets-GEN-AI-Bootcamp/raw/refs/heads/main/Week%203/W3D1%20-%20Data%20Visualization/Student%20Mental%20health.zip"
response = requests.get(url)
with zipfile.ZipFile(io.BytesIO(response.content)) as z:
    csv_name = z.namelist()[0]
    with z.open(csv_name) as f:
        df = pd.read_csv(f, sep=',', engine='python')

# ============================================================
# EXERCISE 4 : Distribution of CGPA (Histogram)
# ============================================================

plt.figure(figsize=(10, 6))
sns.histplot(
    data=df,
    x='What is your CGPA?',
    color='steelblue',
    edgecolor='black',
    shrink=0.8
)
plt.title('Distribution of Students CGPA', fontsize=16, fontweight='bold')
plt.xlabel('CGPA', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print("CGPA Distribution:")
print(df['What is your CGPA?'].value_counts())

# ============================================================
# EXERCISE 5 : Anxiety Levels Across Genders (Bar Plot)
# ============================================================

plt.figure(figsize=(10, 6))
sns.countplot(
    data=df,
    x='Choose your gender',
    hue='Do you have Anxiety?',
    palette='Set2'
)
plt.title('Anxiety Levels Across Different Genders', fontsize=16, fontweight='bold')
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)
plt.legend(title='Has Anxiety?', loc='upper right')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print("\nAnxiety by Gender:")
print(df.groupby('Choose your gender')['Do you have Anxiety?'].value_counts())

# ============================================================
# EXERCISE 6 : Age vs Panic Attacks (Scatter Plot)
# ============================================================

df['Panic Attacks Numeric'] = df['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})

plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x='Age',
    y='Panic Attacks Numeric',
    hue='Do you have Panic attack?',
    palette={'Yes': 'red', 'No': 'steelblue'},
    s=100,
    alpha=0.7
)
plt.title('Relationship Between Age and Panic Attacks', fontsize=16, fontweight='bold')
plt.xlabel('Age', fontsize=12)
plt.ylabel('Panic Attacks (1=Yes, 0=No)', fontsize=12)
plt.yticks([0, 1], ['No (0)', 'Yes (1)'])
plt.legend(title='Panic Attacks')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nPanic Attacks by Age:")
print(df.groupby('Age')['Do you have Panic attack?'].value_counts())
```
