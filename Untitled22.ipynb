```python
# 1. Dépendances de traitement de données et modèles
!pip install -q pandas torch transformers

# 2. Paramétrage des spécifications Serverless
import os
import time
import pandas as pd
from pinecone import Pinecone, ServerlessSpec
from transformers import AutoTokenizer, AutoModel
import torch

cloud = os.getenv('PINECONE_CLOUD', 'aws')
region = os.getenv('PINECONE_REGION', 'us-east-1')

spec = ServerlessSpec(cloud=cloud, region=region)
index_name = 'medical-notes-index'
```

```python
if pc.has_index(name=index_name):
    pc.delete_index(name=index_name)

pc.create_index(
    name=index_name,
    dimension=384,      # Doit correspondre à la taille de sortie de all-MiniLM-L6-v2
    metric='cosine',    # Métrique recommandée pour l'alignement sémantique textuel
    spec=spec
)
```

```python
import requests
import tempfile

with tempfile.TemporaryDirectory() as tmpdirname:
    file_path = os.path.join(tmpdirname, "sample_notes_data.jsonl")

    # URL brute (Raw) pointant vers les données d'exemples de Pinecone
    # Correction: 'master' est souvent remplacé par 'main' dans les dépôts GitHub modernes.
    url = "https://raw.githubusercontent.com/pinecone-io/examples/refs/heads/main/docs/data/sample_notes_data.jsonl"
    response = requests.get(url)
    response.raise_for_status()

    with open(file_path, "wb") as f:
        f.write(response.content)

    df = pd.read_json(file_path, orient='records', lines=True)

# Affichage des dimensions de la table d'entrée
print("Data shape:", df.shape) # Retourne (lignes, colonnes)
display(df.head())
```

```python
# 1. Connexion à notre index nouvellement créé
index = pc.Index(name=index_name)

# Chargement massif à partir de notre DataFrame
index.upsert_from_dataframe(df)

# 2. Boucle de validation de disponibilité
def is_fresh(index):
    stats = index.describe_index_stats()
    vector_count = stats.total_vector_count
    print(f"Vector count currently in index: {vector_count}")
    # On valide dès qu'il y a plus de 0 vecteur présent
    return vector_count > 0

while not is_fresh(index):
    time.sleep(5)

print("\n✅ Index prêt à être interrogé !")
display(index.describe_index_stats())
```

```python


# 1. Fonction de vectorisation locale via PyTorch (Version robuste)
def get_embedding(input_question):
    model_name = 'sentence-transformers/all-MiniLM-L6-v2'
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModel.from_pretrained(model_name)

    # Tokenisation de la question de l'utilisateur
    encoded_input = tokenizer(input_question, padding=True, truncation=True, return_tensors='pt')

    with torch.no_grad():
        model_output = model(**encoded_input)

        # Structure de last_hidden_state : [Batch_Size, Sequence_Length, Hidden_Size]
        # Soit ici pour 1 phrase : [1, Nombre_de_mots, 384]
        token_embeddings = model_output.last_hidden_state

        # On fait la moyenne sur la dimension des mots (dim=1) -> Résultat : [1, 384]
        sentence_embedding = token_embeddings.mean(dim=1)

        # On retire la dimension du batch pour obtenir un vecteur plat -> Résultat : [384]
        embedding = sentence_embedding.squeeze(0)

    return embedding

# 2. Exécution de la recherche vectorielle (Première étape du RAG)
question = "patient has severe chest pain"

# Conversion en liste Python classique (Contiendra exactement 384 nombres)
query_vector = get_embedding(question).tolist()

# Interrogation de l'index Pinecone
# Note : On passe directement 'vector=query_vector' sans les crochets superflus
results = index.query(vector=query_vector, top_k=5, include_metadata=True)

# Tri explicite décroissant basé sur le score de similarité cosinus
sorted_matches = sorted(results['matches'], key=lambda x: x['score'], reverse=True)

print(f" Requête exécutée avec succès ! {len(sorted_matches)} documents récupérés.")
```

```python


# Étape 1 : Affichage des résultats de la recherche vectorielle initiale
def show_results(question, matches):
    print(f"Question: '{question}'")
    print('\nInitial Search Results (Bi-Encoder):')
    for i, match in enumerate(matches):
        print(f'{str(i+1).rjust(4)}. ID: {match["id"]}')
        print(f'   Score: {match["score"]}')
        print(f'   Metadata: {match["metadata"]}')
        print('')

show_results(question, sorted_matches)


# Étape 2 : Préparation des documents (C'est cette variable qui manquait !)
transformed_documents = [
    {
        'id': match['id'],
        # Concaténation propre des métadonnées pour le modèle de reranking
        'reranking_field': '; '.join([f"{key}: {value}" for key, value in match['metadata'].items()])
    }
    for match in results['matches']
]


# Étape 3 : Définition de la requête affinée et exécution du Reranker
refined_query = "patient showing urgent signs of myocardial infarction or severe chest distress"

reranked_results = pc.inference.rerank(
    model="bge-reranker-v2-m3",
    query=refined_query,
    documents=transformed_documents,
    rank_fields=["reranking_field"],
    top_n=3,
    return_documents=True,
)


# Étape 4 : Affichage des résultats finaux reclassés (Syntaxe corrigée ici)
def show_reranked_results(question, matches):
    print(f"Refined Question: '{question}'")
    print('\n Reranked Results (Cross-Encoder):') # Nettoyé du antislash fautif
    for i, match in enumerate(matches):
        print(f'{str(i+1).rjust(4)}. ID: {match.document.id}')
        print(f'   Score: {match.score}')
        print(f'   Reranking Field: {match.document.reranking_field}')
        print('')

# Affichage à partir de l'attribut '.data' fourni par l'API Pinecone
show_reranked_results(refined_query, reranked_results.data)
```

```python
# Suppression de l'index pour éviter tout surcoût inutile
pc.delete_index(name=index_name)
print("Index supprimé avec succès.")
```
