```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from tensorflow.keras import layers, optimizers, losses, metrics

# Configuration globale
num_words = 10000
seed = 42
np.random.seed(seed)

# --- 1. CHARGEMENT ET PRÉTRAITEMENT DES DONNÉES ---
print(" Loading IMDB dataset...")
(train_data, train_labels), (test_data, test_labels) = keras.datasets.imdb.load_data(num_words=num_words)

# Fonction de vectorisation multi-hot (conversion des listes d'entiers en matrices binaires)
def vectorize_sequences(sequences, dimension=10000):
    results = np.zeros((len(sequences), dimension))
    for i, sequence in enumerate(sequences):
        results[i, sequence] = 1.0
    return results

print(" Vectorizing text sequences...")
x_train_full = vectorize_sequences(train_data, dimension=num_words)
x_test = vectorize_sequences(test_data, dimension=num_words)

# Conversion des étiquettes en float32
y_train_full = np.asarray(train_labels).astype('float32')
y_test = np.asarray(test_labels).astype('float32')

# Création de l'ensemble de validation (10 000 exemples pour la validation, le reste pour l'entraînement)
x_val = x_train_full[:10000]
partial_x_train = x_train_full[10000:]

y_val = y_train_full[:10000]
partial_y_train = y_train_full[10000:]

print(f"🔹 Training set shape: {partial_x_train.shape}")
print(f"🔹 Validation set shape: {x_val.shape}")
print(f"🔹 Test set shape: {x_test.shape}\n")


# --- 2. CONSTRUCTION DU MODÈLE BASELINE (20 ÉPOQUES) ---
def build_model():
    model = keras.Sequential([
        layers.Input(shape=(num_words,)),
        layers.Dense(16, activation='relu'),
        layers.Dense(16, activation='relu'),
        layers.Dense(1, activation='sigmoid')
    ])
    model.compile(
        optimizer=optimizers.RMSprop(learning_rate=0.001),
        loss=losses.binary_crossentropy,
        metrics=[metrics.binary_accuracy]
    )
    return model

print(" Training the baseline model for 20 epochs...")
model_baseline = build_model()
history = model_baseline.fit(
    partial_x_train,
    partial_y_train,
    epochs=20,
    batch_size=512,
    validation_data=(x_val, y_val),
    verbose=0 # Mode silencieux pour ne pas surcharger Colab
)
print("Baseline training completed.")


# --- 3. TRACÉ DES COURBES POUR DÉTECTER LE SURAPPRENTISSAGE ---
history_dict = history.history
loss_values = history_dict['loss']
val_loss_values = history_dict['val_loss']
acc = history_dict['binary_accuracy']
val_acc = history_dict['val_binary_accuracy']
epochs = range(1, len(loss_values) + 1)

plt.figure(figsize=(12, 5))

# Graphique des Pertes (Loss)
plt.subplot(1, 2, 1)
plt.plot(epochs, loss_values, 'bo', label='Training loss')
plt.plot(epochs, val_loss_values, 'b', label='Validation loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

# Graphique de la Précision (Accuracy)
plt.subplot(1, 2, 2)
plt.plot(epochs, acc, 'ro', label='Training acc')
plt.plot(epochs, val_acc, 'r', label='Validation acc')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

plt.tight_layout()
plt.show()

# Trouver mécaniquement l'époque optimale (minimum de la perte de validation)
optimal_epochs = np.argmin(val_loss_values) + 1
print(f"\n Visual Analysis: The validation loss is minimal at Epoch {optimal_epochs}.")


# --- 4. RÉENTRAÎNEMENT OPTIMAL ET ÉVALUATION FINALE ---
print(f"\n Retraining a fresh model from scratch for exactly {optimal_epochs} epochs...")
model_final = build_model()
# On réentraîne sur l'intégralité des données d'entraînement disponibles (sans split de validation)
model_final.fit(x_train_full, y_train_full, epochs=optimal_epochs, batch_size=512, verbose=0)

print(" Evaluating final model on the unseen test dataset...")
test_loss, test_acc = model_final.evaluate(x_test, y_test, verbose=0)

print("\n" + "="*40)
print(f" FINAL TEST METRICS (IMDB):")
print(f" - Test Loss: {test_loss:.4f}")
print(f" - Test Accuracy: {test_acc * 100:.2f}%")
print("="*40)
```
