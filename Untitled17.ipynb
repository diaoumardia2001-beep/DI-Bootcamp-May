Task 1: Understanding BERT and XLM-RoBERTa
 Core Architectural Concepts
Before writing code, it is critical to understand the architectural paradigms governing these two landmark Encoder models:

BERT (bert-base-uncased): Monolingual (trained exclusively on English books and Wikipedia). It was pre-trained using two objectives: Masked Language Modeling (MLM) and Next Sentence Prediction (NSP). It uses a WordPiece tokenizer with a concise vocabulary size of ~30,522 tokens. Its default special tokens are [CLS], [SEP], and [PAD].

XLM-RoBERTa (xlm-roberta-base): Multilingual (trained on over 100 languages simultaneously via the CommonCrawl corpus). It builds upon RoBERTa's optimization principles, meaning it dropped the NSP task entirely to train on longer sequences with much larger batch sizes. It implements a SentencePiece (Byte-Pair Encoding) tokenizer, demanding a massive cross-lingual vocabulary size of ~250,002 tokens. Its special tokens are <s>, </s>, and <pad>.

Tasks 2 & 3: Tokenizing Text & Preparing Input Data

```python
import torch
from transformers import AutoTokenizer

# 1. Initialize Tokenizers securely
print(" Loading tokenizers...")
bert_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
xlm_tokenizer = AutoTokenizer.from_pretrained("xlm-roberta-base")

# 2. Define text sequences
sentence_a = "The laboratory analysis confirmed the genomic variations."
sentence_b = "Replication studies are mandatory."
max_length = 16

print("\n=== Tasks 2 & 3: Advanced Encoding via Direct Call ===")

# --- BERT Single Sentence (Direct Call) ---
bert_single = bert_tokenizer(
    sentence_a,
    max_length=max_length,
    padding="max_length",
    truncation=True,
    return_attention_mask=True,
    return_tensors="pt"
)

# --- BERT Sentence Pair (Direct Call) ---
bert_pair = bert_tokenizer(
    sentence_a,
    text_pair=sentence_b,
    max_length=max_length * 2,
    padding="max_length",
    truncation=True,
    return_attention_mask=True,
    return_tensors="pt"
)

# --- XLM-RoBERTa Single Sentence (Direct Call) ---
xlm_single = xlm_tokenizer(
    sentence_a,
    max_length=max_length,
    padding="max_length",
    truncation=True,
    return_attention_mask=True,
    return_tensors="pt"
)

# 3. Print decoded sequences
print(f" Decoded BERT (Single): {bert_tokenizer.decode(bert_single['input_ids'][0])}")
print(f" Decoded BERT (Pair)  : {bert_tokenizer.decode(bert_pair['input_ids'][0])}")
print(f" Decoded XLM-R (Single): {xlm_tokenizer.decode(xlm_single['input_ids'][0])}\n")

print("=== Attention Mask Visualization ===")
print(f"Input Token IDs : {bert_single['input_ids'][0].tolist()}")
print(f"Attention Mask  : {bert_single['attention_mask'][0].tolist()}")
```

```python
import os
import glob
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold

# =====================================================================
# TASK 4: Chargement et Exploration de l'ensemble de données CSV
# =====================================================================
print("=== Task 4: Ingestion et Exploration du Dataset ===")

extraction_target = "/content/dataset_extracted"

# 1. Recherche automatique des fichiers au format .csv extraits du fichier ZIP
csv_files = glob.glob(os.path.join(extraction_target, "**/*.csv"), recursive=True)
if not csv_files:
    raise FileNotFoundError(f"Aucun fichier CSV trouvé dans le dossier {extraction_target}. As-tu bien exécuté l'extraction ZIP à l'étape 1 ?")

# Identification automatique des fichiers d'entraînement (Train) et de test (Test)
train_file = next((f for f in csv_files if "train" in f.lower()), csv_files[0])
test_file = next((f for f in csv_files if "test" in f.lower()), None)

# 2. Lecture des fichiers à l'aide de Pandas
df_train = pd.read_csv(train_file)
print(f" Fichier d'entraînement chargé avec succès : {os.path.basename(train_file)}")
print(f" Forme du Dataset (Lignes, Colonnes) : {df_train.shape}")
print("\n Aperçu des 3 premières lignes de tes données :")
print(df_train.head(3))

# Détection dynamique des colonnes de texte et de catégories (labels)
text_column = 'text' if 'text' in df_train.columns else df_train.columns[0]
label_column = 'label' if 'label' in df_train.columns else df_train.columns[1]

print(f"\n Colonnes détectées pour l'IA : Texte = '{text_column}' | Label = '{label_column}'")

# =====================================================================
# TASK 5: Création des Plis de Validation Croisée (StratifiedKFold)
# =====================================================================
print("\n=== Task 5: Création des Plis (StratifiedKFold) ===")

# Extraction des valeurs sous forme de tableaux Numpy
X = df_train[text_column].values
y = df_train[label_column].values

# Configuration du découpage en 5 parties (Folds)
n_splits = 5
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

fold_train_lists = []
fold_val_lists = []

# Séparation des données en conservant les mêmes proportions de catégories dans chaque pli
for fold_idx, (train_idx, val_idx) in enumerate(skf.split(X, y)):
    train_fold = df_train.iloc[train_idx].reset_index(drop=True)
    val_fold = df_train.iloc[val_idx].reset_index(drop=True)

    # Calcul des pourcentages de chaque classe pour vérifier la répartition exacte
    distribution = val_fold[label_column].value_counts(normalize=True).to_dict()
    dist_str = ", ".join([f"Classe {k}: {v:.1%}" for k, v in distribution.items()])

    print(f" Pli {fold_idx + 1} | Train : {len(train_fold)} lignes | Val : {len(val_fold)} lignes | Ratios de validation -> [{dist_str}]")

    # Stockage dans des listes indépendantes pour ton futur entraînement
    fold_train_lists.append(train_fold)
    fold_val_lists.append(val_fold)

print(f"\n Tout est prêt ! Tes {len(fold_train_lists)} plis stratifiés ont été créés et isolés.")
```
