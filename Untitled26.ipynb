```python


# Partie I : Configuration


# 1. Installation des bibliothèques
!pip install rouge_score==0.1.2
!pip install evaluate
!pip install -U accelerate --quiet
!pip install datasets
!pip install nltk


# 2. Téléchargement des ressources NLTK
import nltk
nltk.download("punkt")
nltk.download("punkt_tab")


# 3. Import des bibliothèques
import pandas as pd
import numpy as np
import torch
from transformers import (
    T5ForConditionalGeneration, AutoTokenizer,
    GPT2LMHeadModel, GPT2Tokenizer
)
import evaluate
import gc
from nltk.tokenize import sent_tokenize
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')

# Vérification du GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Utilisation du device : {device}")


# Partie II : Chargement et exploration du dataset


# Upload des fichiers CSV (train.csv et test.csv)
from google.colab import files
print("Veuillez uploader train.csv et test.csv")
uploaded = files.upload()


# Lecture des fichiers
train_df = pd.read_csv("train.csv")
test_df = pd.read_csv("test.csv")

# Vérification des noms de colonnes – si différents, adaptez ici
print("Colonnes disponibles :", train_df.columns.tolist())
# On suppose que les colonnes s'appellent "prompt_text" et "prompt_title"
# Si ce n'est pas le cas, renommez-les ou modifiez les noms dans la suite.

# Affichage des premières lignes
print("Aperçu de train_df :")
display(train_df.head())
print("Aperçu de test_df :")
display(test_df.head())


# Échantillonnage (100 pour train, 50 pour test)
train_sample = train_df.sample(n=100, random_state=42)
test_sample = test_df.sample(n=50, random_state=42)

# Exploration : premier exemple
print("Premier exemple de l'échantillon d'entraînement :")
print("Article (prompt_text) :")
print(train_sample.iloc[0]["prompt_text"])
print("\nRésumé de référence (prompt_title) :")
print(train_sample.iloc[0]["prompt_title"])

# Affichage des DataFrames échantillonnés
print(f"\nForme de train_sample : {train_sample.shape}")
display(train_sample.head())
print(f"Forme de test_sample : {test_sample.shape}")
display(test_sample.head())


# Partie III : Résumé avec T5


# Générateur de lots
def batch_generator(data, batch_size):
    for i in range(0, len(data), batch_size):
        yield data[i:i+batch_size]


# Fonction summarize_with_t5
def summarize_with_t5(model_name, articles, max_length=150, batch_size=8):
    """
    Génère des résumés pour une liste d'articles avec un modèle T5.
    """
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = T5ForConditionalGeneration.from_pretrained(model_name).to(device)
    model.eval()

    summaries = []
    for batch in tqdm(batch_generator(articles, batch_size),
                      total=(len(articles)+batch_size-1)//batch_size,
                      desc=f"Génération avec {model_name}"):
        # Préfixe "summarize: "
        inputs = tokenizer(
            ["summarize: " + art for art in batch],
            return_tensors="pt",
            truncation=True,
            padding=True,
            max_length=512
        ).to(device)

        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_length=max_length,
                num_beams=4,
                early_stopping=True
            )
        batch_summaries = tokenizer.batch_decode(outputs, skip_special_tokens=True)
        summaries.extend(batch_summaries)

        # Nettoyage mémoire
        torch.cuda.empty_cache()
        gc.collect()

    torch.cuda.empty_cache()
    gc.collect()
    return summaries


# Génération des résumés sur train_sample avec t5-small
train_articles = train_sample["prompt_text"].tolist()
t5_small_summaries = summarize_with_t5("t5-small", train_articles, max_length=150, batch_size=8)


# Affichage des résultats
df_t5_small = train_sample[["prompt_text", "prompt_title"]].copy()
df_t5_small["t5_small_summary"] = t5_small_summaries
print("Résumés générés par t5-small (extrait) :")
display(df_t5_small.head(10))


# Partie IV : Évaluation de l'exactitude (accuracy)


def calculate_accuracy(predictions, references):
    correct = sum(1 for p, r in zip(predictions, references) if p.strip() == r.strip())
    return correct / len(predictions) if predictions else 0.0

accuracy = calculate_accuracy(t5_small_summaries, train_sample["prompt_title"].tolist())
print(f"Exactitude (accuracy) des résumés t5-small : {accuracy:.4f}")

# Discussion
print("\n Discussion : L'exactitude est très faible (souvent nulle) car les résumés générés ne sont jamais identiques mot à mot aux références. Cette métrique n'est pas adaptée pour évaluer la qualité des résumés, car elle est trop stricte et ne tient pas compte des paraphrases ou des variations sémantiques.")


# Partie V : Implémentation de la métrique ROUGE


# Chargement de la métrique ROUGE
rouge_metric = evaluate.load("rouge")


# Prétraitement : ajout de sauts de ligne entre les phrases
def preprocess_for_rouge(text):
    sentences = sent_tokenize(text)
    return "\n".join(sentences)


# Fonction compute_rouge_score
def compute_rouge_score(predictions, references, rouge_types=None):
    """
    Calcule les scores ROUGE (F1) pour des listes de prédictions et de références.
    """
    processed_preds = [preprocess_for_rouge(p) for p in predictions]
    processed_refs = [preprocess_for_rouge(r) for r in references]

    if rouge_types is None:
        rouge_types = ["rouge1", "rouge2", "rougeL"]
    scores = rouge_metric.compute(
        predictions=processed_preds,
        references=processed_refs,
        rouge_types=rouge_types
    )
    # Extraction des valeurs F1
    return {k: v.mid.fmeasure for k, v in scores.items()}

# Test rapide
test_pred = ["This is a summary."]
test_ref = ["This is a summary."]
print("Test ROUGE sur correspondance exacte :")
print(compute_rouge_score(test_pred, test_ref))


# Partie VI : Comprendre les scores ROUGE


# 1. Exact match
print("\n1. Exact match (prédiction = référence)")
pred = ["The cat sat on the mat."]
ref = ["The cat sat on the mat."]
scores = compute_rouge_score(pred, ref)
print(f"ROUGE-1: {scores['rouge1']:.4f}, ROUGE-2: {scores['rouge2']:.4f}, ROUGE-L: {scores['rougeL']:.4f}")


# 2. Prédiction vide
print("\n2. Prédiction vide")
pred = [""]
ref = ["The cat sat on the mat."]
scores = compute_rouge_score(pred, ref)
print(f"ROUGE-1: {scores['rouge1']:.4f}, ROUGE-2: {scores['rouge2']:.4f}, ROUGE-L: {scores['rougeL']:.4f}")


# 3. Effet du stemming
print("\n3. Effet du stemming (racines communes)")
pred = ["The cats sat on the mat."]   # 'cats' vs 'cat'
ref = ["The cat sat on the mat."]
scores = compute_rouge_score(pred, ref)
print(f"Avec 'cats' au lieu de 'cat' : ROUGE-1: {scores['rouge1']:.4f}")  # proche de 1


# 4. Variation de n-grammes
print("\n4. Variation de n-grammes (overlap partiel)")
pred = ["The dog ran on the floor."]
ref = ["The cat sat on the mat."]
scores = compute_rouge_score(pred, ref)
print(f"ROUGE-1: {scores['rouge1']:.4f}, ROUGE-2: {scores['rouge2']:.4f}")


# 5. Symétrie
print("\n5. Symétrie (prédiction <-> référence)")
pred = ["A B C D"]
ref = ["A B C D E"]
scores1 = compute_rouge_score(pred, ref)
scores2 = compute_rouge_score(ref, pred)
print(f"ROUGE-1 (pred->ref): {scores1['rouge1']:.4f}, (ref->pred): {scores2['rouge1']:.4f}")
print("(Les scores sont symétriques car ROUGE utilise la moyenne harmonique F1)")


# Partie VII : Comparaison de petits et grands modèles


# Génération avec t5-base
t5_base_summaries = summarize_with_t5("t5-base", train_articles, max_length=150, batch_size=8)


# Fonction summarize_with_gpt2
def summarize_with_gpt2(model_name, articles, max_length=150, batch_size=8):
    """
    Génère des résumés avec GPT-2 en utilisant le prompt "TL;DR: ".
    """
    tokenizer = GPT2Tokenizer.from_pretrained(model_name)
    model = GPT2LMHeadModel.from_pretrained(model_name).to(device)
    model.eval()

    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token

    summaries = []
    for batch in tqdm(batch_generator(articles, batch_size),
                      total=(len(articles)+batch_size-1)//batch_size,
                      desc=f"Génération avec {model_name}"):
        prompts = ["TL;DR: " + art for art in batch]
        inputs = tokenizer(
            prompts,
            return_tensors="pt",
            truncation=True,
            padding=True,
            max_length=512
        ).to(device)

        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_new_tokens=max_length,
                pad_token_id=tokenizer.eos_token_id,
                do_sample=False,
                num_beams=4
            )
        batch_summaries = tokenizer.batch_decode(outputs, skip_special_tokens=True)
        # Nettoyage : enlever le prompt
        cleaned = []
        for summary, prompt in zip(batch_summaries, prompts):
            if "TL;DR: " in summary:
                summary = summary.split("TL;DR: ")[-1].strip()
            cleaned.append(summary)
        summaries.extend(cleaned)

        torch.cuda.empty_cache()
        gc.collect()

    torch.cuda.empty_cache()
    gc.collect()
    return summaries


# Génération avec GPT-2 (small)
gpt2_summaries = summarize_with_gpt2("gpt2", train_articles, max_length=150, batch_size=8)


# Fonction compute_rouge_per_row
def compute_rouge_per_row(df, summary_col, ref_col="prompt_title"):
    """
    Ajoute des colonnes ROUGE-1, ROUGE-2, ROUGE-L pour chaque ligne.
    Retourne un nouveau DataFrame avec ces colonnes.
    """
    rouge1_list, rouge2_list, rougeL_list = [], [], []
    for idx, row in df.iterrows():
        pred = [row[summary_col]]
        ref = [row[ref_col]]
        scores = compute_rouge_score(pred, ref)
        rouge1_list.append(scores["rouge1"])
        rouge2_list.append(scores["rouge2"])
        rougeL_list.append(scores["rougeL"])

    df_rouge = df.copy()
    df_rouge["rouge1"] = rouge1_list
    df_rouge["rouge2"] = rouge2_list
    df_rouge["rougeL"] = rougeL_list
    return df_rouge


# Construction du DataFrame principal avec les résumés
df_train = train_sample.copy()
df_train["t5_small"] = t5_small_summaries
df_train["t5_base"] = t5_base_summaries
df_train["gpt2"] = gpt2_summaries

# Calcul des scores ROUGE par ligne pour chaque modèle
for model_col in ["t5_small", "t5_base", "gpt2"]:
    df_rouge = compute_rouge_per_row(df_train, model_col)
    # Renommer les colonnes
    df_train[f"{model_col}_rouge1"] = df_rouge["rouge1"]
    df_train[f"{model_col}_rouge2"] = df_rouge["rouge2"]
    df_train[f"{model_col}_rougeL"] = df_rouge["rougeL"]

# Affichage des scores par ligne (extrait)
print("Scores ROUGE par ligne (extrait) :")
display(df_train[["prompt_title", "t5_small_rouge1", "t5_base_rouge1", "gpt2_rouge1"]].head(10))


# Partie VIII : Comparaison de tous les modèles


# Fonction compare_models
def compare_models(df, model_cols, rouge_metrics=["rouge1", "rouge2", "rougeL"]):
    """
    Agrège les scores ROUGE moyens pour chaque modèle.
    Retourne un DataFrame pivot.
    """
    agg = {}
    for model in model_cols:
        for metric in rouge_metrics:
            col = f"{model}_{metric}"
            if col in df.columns:
                agg[(model, metric)] = df[col].mean()
    # Construction du DataFrame
    df_agg = pd.DataFrame(agg.items(), columns=["Model_Metric", "Average"])
    df_agg[["Model", "Metric"]] = pd.DataFrame(df_agg["Model_Metric"].tolist(), index=df_agg.index)
    df_agg = df_agg.pivot(index="Model", columns="Metric", values="Average")
    return df_agg

# Agrégation
model_cols = ["t5_small", "t5_base", "gpt2"]
agg_scores = compare_models(df_train, model_cols)
print("Scores ROUGE moyens par modèle :")
display(agg_scores)


# Fonction compare_models_summaries
def compare_models_summaries(df, model_cols, num_examples=5):
    """
    Affiche côte à côte les résumés générés par chaque modèle pour quelques exemples.
    """
    sample_df = df.sample(n=min(num_examples, len(df)), random_state=42)
    cols = ["prompt_text", "prompt_title"] + model_cols
    display(sample_df[cols])

# Affichage
print("Comparaison des résumés pour quelques articles :")
compare_models_summaries(df_train, model_cols, num_examples=5)


# Soumission
#
#  Lorsque vous avez terminé, sauvegardez ce notebook dans votre Google Drive ou sur GitHub.
# Pour le soumettre, partagez un lien public vers votre Colab :
# - Cliquez sur **Partager** en haut à droite.
# - Modifiez les paramètres pour que "Toute personne ayant le lien" puisse consulter.
# - Copiez le lien et collez-le sur la plateforme DI Learning.
#
#  Optionnel : enregistrez sur GitHub via `Fichier → Enregistrer une copie dans GitHub`.
#
# **Félicitations !** Vous avez réalisé l'évaluation complète de LLM pour la tâche de résumé.
```
