## Exercise 1: Rewrite and Optimize a Vague Prompt
**Scenario**: Manager asks for a LinkedIn post about FlowNest with the vague prompt `"Write something about productivity tips."`

### Tasks
- Identify 3 critical issues with the prompt.
- Rewrite it using: role prompting, clear instructions, and format/tone/length constraints.
- Add a final line to make it reusable for future social posts.

```python
# TODO: list three critical issues with the vague prompt
critical_issues = [
    'No defined role or brand voice — the model has no idea whether to write as a marketer, a wellness coach, or a casual user, so tone consistency is not guaranteed.',
    'No mention of the product (FlowNest) or target audience — the prompt could produce generic tips with zero connection to what is actually being promoted.',
    'No format, length, or platform constraint — LinkedIn posts need a specific structure and length to drive engagement, but the prompt leaves this completely open, risking an unusable wall of text.',
]
critical_issues
```

```python
# TODO: rewrite the prompt using role, instructions, and constraints
rewritten_prompt = (
    "Act as a social media copywriter for a productivity startup. "
    "Write a LinkedIn post promoting our new focus app, FlowNest, targeting "
    "busy professionals who struggle with distractions. Give 3 actionable "
    "productivity tips, each naturally tied to a FlowNest feature. Use short "
    "paragraphs or bullet points, an encouraging and professional tone, and "
    "keep the total post under 150 words. End with a soft call-to-action to "
    "try the app. Use this format whenever we promote productivity tools on "
    "social media."
)
rewritten_prompt
```

## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
# TODO: write the full multi-part prompt
quiz_prompt = (
    "Act as a friendly middle school science teacher. Based on the attached "
    "article about volcanic eruptions, do the following: "
    "(1) Summarize the article in exactly 2 simple bullet points, using "
    "vocabulary appropriate for 11-13-year-olds. "
    "(2) Generate 3 multiple-choice questions about the article, each with "
    "1 correct answer and 2 plausible distractors, written in a friendly and "
    "encouraging tone. "
    "(3) Format the output as slide-ready content, with one summary slide "
    "and one question per slide, clearly labeled 'Slide 1', 'Slide 2', etc., "
    "suitable for pasting into Google Slides."
)
quiz_prompt
```

```python
# TODO: explain three improvements over the vague prompt
improvements = [
    'Defines a clear role and audience (a science teacher writing for 11-13-year-olds), which controls vocabulary level and tone instead of leaving them to guess.',
    'Specifies an exact structure (2 summary bullet points, 3 multiple-choice questions with 1 correct answer and 2 distractors each), preventing an inconsistent or incomplete quiz.',
    'States the required output format (slide-ready content labeled Slide 1, Slide 2, etc.), ensuring the result can be pasted directly into Google Slides instead of needing manual reformatting.',
]
improvements
```
