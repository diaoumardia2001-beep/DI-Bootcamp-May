```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

temperatures = np.random.uniform(low=-5, high=35, size=(10, 12))

cities = [f"City {i+1}" for i in range(10)]
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

df_weather = pd.DataFrame(data=temperatures, index=cities, columns=months)
df_weather['Annual_Avg'] = df_weather.mean(axis=1)


hottest_city = df_weather['Annual_Avg'].idxmax()
hottest_temp = df_weather['Annual_Avg'].max()

coldest_city = df_weather['Annual_Avg'].idxmin()
coldest_temp = df_weather['Annual_Avg'].min()

print(f"Hottest City: {hottest_city} ({hottest_temp:.2f}°C)")
print(f"Coldest City: {coldest_city} ({coldest_temp:.2f}°C)")
months_only = df_weather.drop(columns=['Annual_Avg'])


plt.figure(figsize=(11, 5.5))


for city in months_only.index:
    if city in ["City 7", "City 10"]:
        # Thick dashed lines for highlight cities
        plt.plot(months, months_only.loc[city], label=city, 
                color='#16a085', linestyle='--', linewidth=2.5)
    else:
        # Sleek, clean lines for general trends
        plt.plot(months, months_only.loc[city], label=city, 
                color='#5da994', alpha=0.8, linewidth=1.5)


plt.title('Monthly Temperature Trends for Each City', fontsize=14, pad=15, color='#2c3e50')
plt.xlabel('Month', fontsize=11, labelpad=8)
plt.ylabel('Temperature (°C)', fontsize=11, labelpad=8)


plt.grid(True, linestyle='--', alpha=0.5, color='#d1d8e0')
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#2c3e50')
ax.spines['bottom'].set_color('#2c3e50')


plt.legend(loc='lower left', fontsize=9, framealpha=0.9)

plt.tight_layout()
plt.show()
```
