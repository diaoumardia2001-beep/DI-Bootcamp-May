```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
nom_colonnes = ['Exam_1_Score', 'Exam_2_Score', 'Admitted']
df = pd.read_csv("/content/ex2data1.txt", header=None, names=nom_colonnes)

print("--- Aperçu des 5 premières lignes ---")
print(df.head())
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Exam_1_Score', y='Exam_2_Score', hue='Admitted',
                style='Admitted', palette="Set1", s=100)

plt.title("University Admission based on Exam Scores (ex2data1)")
plt.xlabel("Exam 1 Score")
plt.ylabel("Exam 2 Score")
plt.legend(title="Admission Status", labels=["Not Admitted (0)", "Admitted (1)"])
plt.grid(True)
plt.show()
X = df[['Exam_1_Score', 'Exam_2_Score']]
y = df['Admitted']
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
model = LogisticRegression(random_state=42)
model.fit(X_scaled, y)

print("\n--- Entraînement réussi ---")
print(f"Intercept (b) : {model.intercept_[0]:.4f}")
print(f"Coefficients (w1, w2) : {model.coef_[0]}")
y_pred = model.predict(X_scaled)
accuracy = accuracy_score(y, y_pred)

print(f"\n Score de Précision Final (Accuracy) : {accuracy:.4f} ({accuracy * 100:.2f}%)")
```

Model Evaluation & Interpretation
Data Observations: The scatter plot reveals that students with a combined low score on both exams fail to get admitted. There is a diagonal trend showing that higher scores in both exams strongly guarantee admission.

Model Performance: Our Logistic Regression model achieved an accuracy of 89.00% on this dataset (ex2data1.txt). This means the linear decision boundary correctly classifies 89 out of 100 students based solely on their two exam results.

Conclusion: Logistic Regression is an efficient and highly interpretable model for this classification task.
