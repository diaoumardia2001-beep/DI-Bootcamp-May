```python
# TODO: capture definitions and key characteristics
open_source_levels = {
    "Fully Open": {
        "definition": "A model where the architecture, pretrained weights, training code, and often the training data or data recipe are all publicly released, allowing complete transparency and control.",
        "what_is_open": ["Model architecture and code", "Pretrained weights and (often) training data/recipe"],
        "what_you_can_do": ["Inspect the full model internals and training process", "Retrain, fine-tune, or modify the model from scratch if needed"],
        "what_you_cannot_do": ["N/A in most cases — this is the most permissive level", "May still be bound by a specific license's attribution or use restrictions, if any apply"],
    },
    "Weights Released": {
        "definition": "A model where only the pretrained weights are released publicly, without the original training code, data pipeline, or full reproducibility details.",
        "what_is_open": ["Pretrained model weights", "Usually the model architecture/config (needed to load the weights)"],
        "what_you_can_do": ["Fine-tune the model on your own data", "Run inference locally or deploy it without depending on the original creator's infrastructure"],
        "what_you_cannot_do": ["Fully audit exactly how the model was originally trained", "Reproduce the base model from scratch (training data/code isn't public)"],
    },
    "Architecture Only": {
        "definition": "Only the model's architectural design (layer structure, configuration, sometimes a research paper) is published, without any pretrained weights.",
        "what_is_open": ["Model architecture/design", "Configuration details and often the accompanying research paper"],
        "what_you_can_do": ["Study and reproduce the model design", "Implement and pretrain your own version if you have the compute budget"],
        "what_you_cannot_do": ["Use the model immediately — there are no usable pretrained weights", "Fine-tune without first completing a full (often costly) pretraining run"],
    },
}
open_source_levels
```

```python
# TODO: fill the comparison table (keep markdown formatting)
comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | Architecture, weights, training code, and often training data/recipe | You can inspect, modify, and retrain the entire model end-to-end |
| Weights Released | Pretrained weights and model architecture/config | You can fine-tune and deploy, but can't audit or reproduce the original training process |
| Architecture Only | Model architecture/design and configuration, sometimes a research paper | You know the structure but must pretrain from scratch to get a usable model |"""
print(comparison_table)
```

```python
# TODO: 3-5 sentence comparative paragraph
comparative_paragraph = """Fully Open models release everything — architecture, weights, and often training data — giving developers complete transparency and the ability to inspect, retrain, or audit the model, though this level of openness is rare for large frontier models. Weights Released models, like Llama 2 or Mistral, let you download and fine-tune a pretrained model for your own use case, but you can't fully see or reproduce how it was originally trained, since the training code and data pipeline stay private. Architecture Only releases share the model's design through papers or configuration files without providing any pretrained weights, which lets researchers study or reproduce the approach but leaves you without a usable model unless you can afford a full pretraining run yourself. In short, the more open the release, the more control and auditability you gain, but truly Fully Open large models remain the exception rather than the norm."""
print(comparative_paragraph)
```
