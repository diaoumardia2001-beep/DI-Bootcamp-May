What You Need to Do
1. Load and Preprocess the MNIST Dataset

Load the MNIST dataset using TensorFlow/Keras
Normalize the image pixel values to be between 0 and 1

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
from sklearn.metrics import confusion_matrix
import seaborn as sns

# 1. Load the dataset
(X_train_raw, y_train_raw), (X_test_raw, y_test_raw) = mnist.load_data()

# 2. Normalize pixel intensities from [0, 255] to [0.0, 1.0]
X_train = X_train_raw.astype('float32') / 255.0
X_test = X_test_raw.astype('float32') / 255.0

# 3. One-hot encode the target labels
y_train_oh = to_categorical(y_train_raw, num_classes=10)
y_test_oh = to_categorical(y_test_raw, num_classes=10)

# 4. Display sample images with their actual digit labels
plt.figure(figsize=(10, 3))
for i in range(5):
    plt.subplot(1, 5, i + 1)
    plt.imshow(X_train_raw[i], cmap='gray')
    plt.title(f"Label: {y_train_raw[i]}")
    plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Define the sequential model structure
model = models.Sequential([
    # Input translation: Flattens 2D matrices (28x28) into 1D arrays (784 features)
    layers.Flatten(input_shape=(28, 28)),

    # Hidden Layer 1: 128 neurons utilizing ReLU to avoid gradient saturation
    layers.Dense(128, activation='relu'),

    # Hidden Layer 2: 64 neurons to extract refined sub-features
    layers.Dense(64, activation='relu'),

    # Output Layer: 10 neurons mapping raw scores to a probability distribution
    layers.Dense(10, activation='softmax')
])

# Compile configuration
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
# Fit the model to training observations
history = model.fit(
    X_train,
    y_train_oh,
    epochs=10,
    batch_size=64,
    validation_split=0.1  # Tracks performance trends on an unseen 10% slice
)
```

```python
# Plot Training & Validation performance
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Loss trends
ax1.plot(history.history['loss'], label='Train Loss')
ax1.plot(history.history['val_loss'], label='Val Loss')
ax1.set_title('Loss Decay Curve')
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Loss')
ax1.legend()

# Accuracy trends
ax2.plot(history.history['accuracy'], label='Train Accuracy')
ax2.plot(history.history['val_accuracy'], label='Val Accuracy')
ax2.set_title('Accuracy Evaluation Curve')
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Accuracy')
ax2.legend()

plt.show()
```

```python
# 1. Compute baseline test dataset metric
test_loss, test_acc = model.evaluate(X_test, y_test_oh, verbose=0)
print(f"🎯 Final Test Dataset Accuracy: {test_acc * 100:.2f}%\n")

# 2. Extract model predictions
y_prob = model.predict(X_test)
y_pred = np.argmax(y_prob, axis=1)

# 3. Build Confusion Matrix
cm = confusion_matrix(y_test_raw, y_pred)

# 4. Display heatmap using Seaborn
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=range(10), yticklabels=range(10))
plt.title('MNIST Error Distribution (Confusion Matrix)')
plt.xlabel('Predicted Digit Label')
plt.ylabel('True Digit Label')
plt.show()
```

The Main Diagonal: The dark blue boxes stretching diagonally from top-left to bottom-right show correct predictions.

Off-Diagonal Clusters: Look for values outside the main diagonal line. For example, you will likely notice clusters around True 4 misclassified as 9, or True 7 misclassified as 9 / 2. These patterns occur because handwritten digits share similar topological strokes (e.g., the vertical and circular loops of a 4 look like a 9 if the top loop is compressed).
